# 2차시 강사용 Notebook — 리스트와 딕셔너리

학생용 Notebook과 동일한 흐름이며, 예상 결과와 설명 포인트를 함께 담았습니다.

## 1단계. 리스트 만들기
**설명 포인트**: 1차시의 변수 하나(`pass_count = 34`)와 비교해 "여러 값을 한 번에 담을 수 있다"는 점을 강조.

```python
temperatures = [295.4, 298.1, 312.6, 301.0, 296.8]
print(temperatures)
```

## 2단계. 인덱싱으로 값 꺼내기
**어려워할 부분**: 인덱스가 0부터 시작한다는 점. `temperatures[1]`이 '두 번째' 값임을 직접 실행해서 보여준다.

```python
print(temperatures[0])
print(temperatures[2])
print(temperatures[-1])
```

## 3단계. 슬라이싱
**시연 방법**: `temperatures[1:3]`과 `temperatures[1:4]`를 나란히 실행해 끝 인덱스가 포함되지 않음을 비교.

```python
print(temperatures[1:3])
print(temperatures[:2])
print(temperatures[2:])
```

## 4단계. 길이와 추가

```python
print(len(temperatures))
temperatures.append(303.2)
print(temperatures)
print(len(temperatures))
```

## 5단계. 최댓값·최솟값·평균 계산하기
**설명 포인트**: `round()`로 소수점을 정리하고 1차시의 f-string을 복습으로 다시 사용.

```python
highest = max(temperatures)
lowest = min(temperatures)
average = sum(temperatures) / len(temperatures)

print(f"최댓값: {highest}, 최솟값: {lowest}, 평균: {round(average, 1)}")
```

## 6단계. 딕셔너리 만들기
**설명 포인트**: 리스트와 곧바로 비교 — "순서 대신 이름표(키)로 값을 구분한다".

```python
equipment = {
    "설비번호": "EQ-01",
    "공정명": "식각",
    "온도_섭씨": 301.2,
}
print(equipment)
```

## 7단계. 키로 값 접근하기
**어려워할 부분**: 없는 키에 접근하면 `KeyError`가 난다는 점을 실제로 `equipment["압력_Pa"]`를 실행해 보여주면 효과적.

```python
print(equipment["설비번호"])
print(equipment["온도_섭씨"])
```

## 8단계. 값 수정하기 / 새 키 추가하기

```python
equipment["온도_섭씨"] = 305.5
equipment["압력_Pa"] = 1012.3
print(equipment)
```

## 9단계. 리스트 안에 딕셔너리 담기
**설명 포인트**: 이번 차시의 핵심 도약 지점. 4차시 Pandas의 표(행=딕셔너리, 표=리스트) 개념과 미리 연결해 언급한다.

```python
equipment_list = [
    {"설비번호": "EQ-01", "공정명": "식각", "온도_섭씨": 301.2},
    {"설비번호": "EQ-02", "공정명": "증착", "온도_섭씨": 308.7},
    {"설비번호": "EQ-03", "공정명": "포토", "온도_섭씨": 297.5},
]

print(len(equipment_list))
print(equipment_list[0])
print(equipment_list[0]["온도_섭씨"])
print(equipment_list[1]["설비번호"])
```

## 10단계(도전). 종합 실습 — 나만의 설비 목록 만들기
**시연 방법**: 학생 한 명에게 설비 이름을 즉석에서 물어보고 함께 채워 넣으면 참여도가 높아진다.

```python
my_equipment = [
    {"설비번호": "EQ-04", "공정명": "세정", "온도_섭씨": 299.0},
    {"설비번호": "EQ-05", "공정명": "산화", "온도_섭씨": 302.4},
]

print(f"{my_equipment[0]['설비번호']} 설비는 {my_equipment[0]['공정명']} 공정입니다.")
print(f"{my_equipment[1]['설비번호']} 설비는 {my_equipment[1]['공정명']} 공정입니다.")
```

## 11단계. 오늘의 학습을 한 문장으로 정리하기
**진행 방법**: 학생 두세 명에게 자기 문장을 소리 내어 말하게 한다. 문장이 나오지 않는 학생에게는 "리스트와 딕셔너리는 값을 무엇으로 구분하나요?"만 물어보면 된다.
**설명 포인트**: 결과 해석 문장 쓰기는 1차시부터 10차시까지 매 차시 빠지지 않는 고정 활동이다(docs/curriculum.md 「차시 간 연결 원칙」). 코드를 친 것으로 끝내지 않고 자기 말로 바꿔 말해보게 하는 것이 이 과정의 목표 — "데이터로 공정 상태를 설명할 수 있는 사람" — 에 직접 닿는 활동이므로 시간이 모자라도 2~3분은 반드시 확보한다.

> (예시 답안) 리스트는 순서로 값을 구분하고, 딕셔너리는 이름표(키)로 값을 구분한다. 리스트 안에 딕셔너리를 담으면 여러 설비 정보를 한 번에 표현할 수 있다.

## 오류 대처 방법
- `IndexError` 발생 시: `len()`으로 리스트 길이를 먼저 확인.
- `KeyError` 발생 시: `print(딕셔너리)`로 전체 키를 먼저 확인.

## 확장 실습(빠른 학습자용)
온도만 모은 새 리스트를 만들어 `max()`를 적용해보게 하면 3차시(반복문으로 자동화) 예고가 된다.

```python
temps_only = [equipment_list[0]["온도_섭씨"], equipment_list[1]["온도_섭씨"], equipment_list[2]["온도_섭씨"]]
print(temps_only)
print(max(temps_only))
```

## 12단계. AI에게 질문하며 더 알아보기
**설명 포인트**: 리스트/튜플, 딕셔너리 키 제약 같은 질문은 4차시 Pandas의 인덱스 개념과도
연결되니 답변을 들을 때 미리 짚어주면 좋다.

질문 예시:
- "파이썬 리스트와 튜플은 어떻게 다른가요?"
- "딕셔너리에서 키(key)로 쓸 수 있는 값과 쓸 수 없는 값은 무엇인가요?"
- "딕셔너리 안에 리스트를 담을 수도 있나요?"
- "리스트의 append()와 insert()는 어떻게 다른가요?"

## 13단계. AI에게 코드 생성 요청하고 직접 실행해보기
**설명 포인트**: `max(..., key=lambda ...)`는 오늘 배우지 않은 문법이다 — "AI가 준 코드도 이해하고
써야 한다"는 메시지를 전달하기 좋은 타이밍이다.

프롬프트 예시: "설비 정보 딕셔너리가 여러 개 담긴 리스트에서, 온도가 가장 높은 설비 하나만
찾아주는 코드를 만들어줘."

아래는 AI가 생성해줄 수 있는 예시 코드와 실행 결과다.

```python
hottest = max(equipment_list, key=lambda eq: eq["온도_섭씨"])
print(f"가장 온도가 높은 설비는 {hottest['설비번호']}이고, 온도는 {hottest['온도_섭씨']}도입니다.")
```

## 14단계. 연습문제 — 압력 리스트와 새 설비 딕셔너리
**설명 포인트**: 7단계(1차시 온도·설비 문장 만들기)와 같은 목적의 단계다. 오늘 배운 리스트
생성·통계 계산(`max`/`min`/`sum`/`len`)과 딕셔너리 키 추가를 새로운 데이터(압력)로 한 번 더
반복 연습시킨다. 빠른 학습자에게는 도전(새 설비 딕셔너리)까지, 느린 학습자에게는 리스트 통계
계산까지만 요구해도 충분하다.

```python
pressures = [1010.2, 1005.4, 1022.9, 998.7, 1011.0]

highest_p = max(pressures)
lowest_p = min(pressures)
average_p = sum(pressures) / len(pressures)

print(f"최댓값: {highest_p}, 최솟값: {lowest_p}, 평균: {round(average_p, 1)}")
```

### 도전. 새 설비 딕셔너리에 압력 추가하기
**시연 방법**: 방금 계산한 `average_p`를 그대로 새 딕셔너리의 값으로 재사용하는 것을 보여주면,
"한 계산 결과를 다른 자료구조에 다시 담을 수 있다"는 감각을 심어주기 좋다.

```python
eq04 = {
    "설비번호": "EQ-04",
    "공정명": "세정",
    "온도_섭씨": 299.9,
}

eq04["압력_Pa"] = average_p
print(eq04)
```

## 15단계. AI로 재미있는 미니 프로그램 만들기 🎉 — 오늘의 럭키 설비 뽑기
**설명 포인트**: 8단계(1차시 공정 운세 뽑기)와 같은 목적의 마무리 단계다. 정확한 코드 자체보다
"오늘 만든 리스트·딕셔너리를 AI와 함께 재미있는 프로그램으로 확장해보는 경험"을 즐기도록 유도한다.

프롬프트 예시: "설비 정보 딕셔너리가 여러 개 담긴 리스트(equipment_list)에서, random.choice()로
하나를 무작위로 뽑아서 '오늘의 럭키 설비'라는 제목과 함께 설비번호·공정명·온도를 예쁘게 출력하는
파이썬 코드를 만들어줘."

**주의 안내**: `random` 모듈은 오늘 처음 보는 표현이다 — 리스트·딕셔너리는 오늘 배운 내용이니,
`random.choice()`가 리스트에서 하나를 무작위로 골라준다는 정도만 짚어주면 충분하다.

아래는 AI가 생성해줄 수 있는 예시 코드다(자동 실행 검증에서 결과가 항상 같도록 `random.seed(7)`을
사용했다 — 수업 시연에서는 seed를 빼서 매번 다른 설비가 뽑히게 해도 좋다). 앞서 만든 `eq04`를
`equipment_list`에 더해 네 개 설비 중에서 뽑히도록 구성했다.

```python
import random

random.seed(7)  # 수업 시연에서는 seed를 빼서 매번 다른 결과가 나오게 해도 좋습니다.

equipment_list = [
    {"설비번호": "EQ-01", "공정명": "식각", "온도_섭씨": 301.2},
    {"설비번호": "EQ-02", "공정명": "증착", "온도_섭씨": 308.7},
    {"설비번호": "EQ-03", "공정명": "포토", "온도_섭씨": 297.5},
    eq04,
]

lucky = random.choice(equipment_list)
print("🎉 오늘의 럭키 설비 🎉")
print(f"설비번호: {lucky['설비번호']}")
print(f"공정명: {lucky['공정명']}")
print(f"온도: {lucky['온도_섭씨']}도")
```
