## 1단계. 리스트 만들기
**설명 포인트**: 1차시의 변수 하나(`pass_count = 32`)와 비교해 "여러 값을 한 번에 담을 수 있다"는 점을 강조.

**진행 방법**: 강의 자료 순서대로 과일 리스트(`fruits`) 세 셀로 리스트와 인덱스의 감을 잡은 뒤 온도 리스트로 넘어간다.

**예상 결과**: `['사과', '바나나', '포도']` → `사과` → `포도` → `[295.4, 298.1, 312.6, 301.0, 296.8]`

### 따라 해보기 — 아주 간단한 리스트
반도체 데이터로 바로 들어가기 전에, 과일 이름 세 개를 담은 가장 단순한 리스트로 감을 잡습니다
(강의 자료 "먼저 아주 간단한 예로 감을 잡자"). 완성된 코드이니 한 셀씩 실행해 결과를 확인해 보세요.

```python
# 과일 이름 3개를 담은 리스트를 만들어 fruits에 저장하고, 리스트 전체를 출력합니다.
# 실행하면 ['사과', '바나나', '포도']가 출력됩니다 — 대괄호와 콤마까지 그대로 보입니다.
# 리스트: 대괄호 [ ] 안에 값을 콤마(,)로 구분해 나열하면 여러 값을 순서대로 한 번에 담을 수 있습니다.
fruits = ["사과", "바나나", "포도"]
print(fruits)
```

```python
# 리스트에서 첫 번째 값을 꺼내 출력합니다.
# 실행하면 사과가 출력됩니다. 첫 번째 값의 순서 번호(인덱스)는 1이 아니라 0입니다.
# 리스트이름[번호]: 대괄호 안에 인덱스를 넣으면 그 위치의 값을 꺼냅니다.
print(fruits[0])    # 첫 번째 값
```

```python
# 리스트에서 마지막 값을 꺼내 출력합니다.
# 실행하면 포도가 출력됩니다. -1처럼 음수 인덱스는 뒤에서부터 셉니다.
print(fruits[-1])   # 마지막 값
```

### 반도체 데이터로 — 온도 리스트 만들기
리스트는 글자뿐 아니라 숫자도 같은 방식으로 담을 수 있습니다. 이제 한 설비의 온도 측정값 5개를 리스트에 담습니다.

```python
# 한 설비에서 측정한 온도 값 5개를 리스트로 만들어 temperatures에 저장하고 출력합니다.
# 실행하면 [295.4, 298.1, 312.6, 301.0, 296.8]이 출력됩니다. 숫자도 글자와 같은 방식으로 리스트에 담습니다.
temperatures = [295.4, 298.1, 312.6, 301.0, 296.8]
print(temperatures)
```

## 2단계. 인덱싱으로 값 꺼내기
**어려워할 부분**: 인덱스가 0부터 시작한다는 점. `temperatures[1]`이 '두 번째' 값임을 직접 실행해서 보여준다.

마지막 셀(`temperatures[1]`)이 강의 "가장 흔한 함정" 확인용이다 — 298.1(두 번째 값)이 나오는 것을 보여준다.

**예상 결과**: `295.4` → `312.6` → `296.8` → `298.1`

```python
# temperatures에서 첫 번째 값을 꺼내 출력합니다.
# 실행하면 295.4가 출력됩니다.
print(temperatures[0])    # 첫 번째 값
```

```python
# temperatures에서 세 번째 값을 꺼내 출력합니다.
# 실행하면 312.6이 출력됩니다. 인덱스가 0부터 시작하므로 세 번째 값의 인덱스는 3보다 하나 작습니다.
print(temperatures[2])    # 세 번째 값
```

```python
# temperatures에서 마지막 값을 꺼내 출력합니다.
# 실행하면 296.8이 출력됩니다. 값이 몇 개인지 몰라도 음수 인덱스로 뒤에서부터 꺼낼 수 있습니다.
print(temperatures[-1])   # 마지막 값(음수 인덱스는 뒤에서부터)
```

```python
# [가장 흔한 함정 확인] 인덱스 1은 "첫 번째"가 아니라 "두 번째" 값입니다.
# 실행하면 298.1(두 번째 값)이 출력됩니다 — 첫 번째 값 295.4가 아닌 것을 눈으로 확인해 보세요.
print(temperatures[1])
```

## 3단계. 슬라이싱
**시연 방법**: `temperatures[1:3]`과 `temperatures[1:4]`를 나란히 실행해 끝 인덱스가 포함되지 않음을 비교.

**예상 결과**: `[298.1, 312.6]` → `[295.4, 298.1]` → `[312.6, 301.0, 296.8]` → (직접 해보기) `[301.0, 296.8]`

```python
# 슬라이싱으로 두 번째·세 번째 값(298.1, 312.6)만 잘라 출력합니다.
# 실행하면 [298.1, 312.6]이 출력됩니다 — 결과도 리스트라서 대괄호로 감싸여 나옵니다.
# 리스트이름[시작:끝]: 시작 인덱스부터 끝 인덱스 "바로 앞"까지 잘라냅니다. 끝 인덱스 자신은 들어가지 않습니다.
print(temperatures[1:3])   # 인덱스 1~2 (3은 포함 안 됨)
```

```python
# 시작 인덱스를 비워 두면 처음부터 잘라냅니다. 인덱스 0, 1의 값을 출력합니다.
# 실행하면 [295.4, 298.1]이 출력됩니다.
print(temperatures[:2])    # 처음부터 인덱스 1까지
```

```python
# 끝 인덱스를 비워 두면 마지막 값까지 잘라냅니다. 인덱스 2부터 끝까지 출력합니다.
# 실행하면 [312.6, 301.0, 296.8]이 출력됩니다.
print(temperatures[2:])    # 인덱스 2부터 끝까지
```

### 직접 해보기 — 음수 인덱스와 슬라이싱 함께 쓰기
강의 자료 「슬라이싱 — 여러 값을 한 번에 잘라내기」 절의 "직접 해보기" 상자를 노트북에서 실행해 봅니다.

```python
# [직접 해보기] 음수 인덱스와 슬라이싱을 함께 쓰면 어떤 값이 나오는지 확인합니다.
# 실행하면 리스트가 출력됩니다 — 원래 temperatures의 어느 부분이 잘려 나왔는지 비교해 보세요.
print(temperatures[-2:])
```

> **참고**
>
> 강의 자료 「슬라이싱 — 여러 값을 한 번에 잘라내기」 절의 "직접 해보기" 상자에 이 셀의 코드가 그대로 적혀 있습니다. 음수 인덱스는 뒤에서부터 세는 번호이고(「인덱싱 — 순서로 값 꺼내기」 절), `[시작:]`처럼 끝 인덱스를 비워 두면 시작 위치부터 마지막 값까지 잘라냅니다.
>
> 정답 해설: 빈칸은 `-2`. `temperatures[-2:]`는 "뒤에서 두 번째 값부터 끝까지"라서 마지막 두 값이 리스트로 나온다. 음수 인덱스와 슬라이싱을 함께 쓰면 리스트 길이를 몰라도 끝부분을 잘라낼 수 있다.
>
> 예상 결과: `[301.0, 296.8]` (4단계 `append()` 전이라 값 5개 기준. `append()` 뒤에 다시 실행하면 `[296.8, 303.2]`)

## 4단계. 길이와 추가

**주의**: `append()` 셀을 두 번 실행하면 303.2가 두 번 들어가 5단계 결과가 달라진다. 강의 체크포인트 "append()는 아무것도 돌려주지 않는다"(`temperatures = temperatures.append(303.2)` 금지)와 함께 짚는다.

**예상 결과**: `5` → `[295.4, 298.1, 312.6, 301.0, 296.8, 303.2]` → `6`

```python
# 리스트에 값이 몇 개 들어 있는지 세어 출력합니다.
# 실행하면 5가 출력됩니다.
# len(리스트): 리스트에 담긴 값의 개수(길이)를 돌려주는 함수입니다.
print(len(temperatures))     # 리스트 길이(값 개수)
```

```python
# 리스트 끝에 새 온도 값 303.2를 추가한 뒤, 리스트 전체를 다시 출력합니다.
# 실행하면 맨 끝에 303.2가 붙은 6개짜리 리스트가 출력됩니다. 이 셀은 한 번만 실행하세요(다시 실행하면 303.2가 또 붙습니다).
# 리스트.append(값): 리스트 끝에 값을 하나 붙입니다. 돌려주는 결과가 없으므로 temperatures = ...처럼 다시 대입하지 않습니다.
temperatures.append(303.2)   # 끝에 새 값 추가
print(temperatures)
```

```python
# 값을 하나 추가한 뒤 리스트 길이를 다시 셉니다.
# 실행하면 6이 출력됩니다 — 처음 5개에서 1개 늘었습니다. 7 이상이 나오면 위 셀을 여러 번 실행한 것이니 1단계부터 다시 실행하세요.
print(len(temperatures))
```

## 5단계. 최댓값·최솟값·평균 계산하기
**설명 포인트**: `round()`로 소수점을 정리하고 1차시의 f-string을 복습으로 다시 사용.

**예상 결과**(4단계에서 303.2를 추가한 6개 값 기준): `312.6` → `295.4` → `1807.1` → `평균 온도는 301.2도입니다.` → (직접 해보기) `17.200000000000045` → `온도 범위는 17.2도입니다.` 소수점 아래가 길게 나오는 이유(부동소수 오차)와 `:.1f` 서식으로 정리하는 방법을 함께 짚는다.

```python
# 리스트에서 가장 큰 값을 찾아 출력합니다.
# 실행하면 312.6이 출력됩니다.
# max(리스트) / min(리스트) / sum(리스트): 리스트 전체의 최댓값 / 최솟값 / 합계를 돌려주는 파이썬 내장 함수입니다.
print(max(temperatures))     # 최댓값
```

```python
# 리스트에서 가장 작은 값을 찾아 출력합니다.
# 실행하면 295.4가 출력됩니다.
print(min(temperatures))     # 최솟값
```

```python
# 리스트의 모든 값을 더한 합계를 출력합니다.
# 실행하면 1807.1이 출력됩니다(4단계에서 추가한 303.2까지 6개 값의 합).
print(sum(temperatures))     # 합계
```

```python
# 평균(합계 ÷ 개수)을 계산해 average에 저장하고, 소수 첫째 자리까지 반올림해 문장으로 출력합니다.
# 실행하면 "평균 온도는 301.2도입니다."가 출력됩니다.
# round(값, 1): 값을 소수 첫째 자리까지 반올림합니다.
average = sum(temperatures) / len(temperatures)
print(f"평균 온도는 {round(average, 1)}도입니다.")   # round()로 소수점 정리 + 1차시 f-string 복습
```

### 직접 해보기 — 온도 범위 계산하기
강의 자료 「리스트에서 최댓값·최솟값·평균 구하기」 절의 "직접 해보기" 상자입니다. 최고 온도와 최저 온도의 차이(온도 범위)를 계산하고 문장으로 출력합니다.

```python
# [직접 해보기] 온도 범위(최고 온도 - 최저 온도)를 계산해 temp_range에 저장하고, 값을 출력해 확인합니다.
# 실행하면 17.2에 아주 가까운 값이 소수점 아래 여러 자리까지 길게 나옵니다(컴퓨터가 소수를 2진수로 계산하며 생기는 아주 작은 오차).
temp_range = max(temperatures) - min(temperatures)
print(temp_range)  # (중간 확인)
```

> **참고**
>
> 강의 자료 「리스트에서 최댓값·최솟값·평균 구하기」 절의 "직접 해보기" 상자에 이 계산식이 그대로 있습니다. 같은 절의 코드 블록에서 리스트 전체의 최댓값과 최솟값을 구한 함수 두 개를 다시 보세요 — 두 결과를 빼면 온도 범위가 됩니다.
>
> 정답 해설: 빈칸은 차례로 `max`, `min`. `max(temperatures) - min(temperatures)`는 312.6 - 295.4다. 출력에 소수점 아래 자리가 길게 붙는 것은 컴퓨터가 소수를 2진수로 저장하며 생기는 아주 작은 오차이고, 다음 셀의 서식 기호로 정리한다.
>
> 예상 결과: `17.200000000000045` (4단계에서 303.2를 추가한 6개 값 기준. 최댓값 312.6, 최솟값 295.4)

```python
# [직접 해보기] 온도 범위를 소수 첫째 자리까지만 보이게 해서 문장으로 출력합니다.
# 실행하면 "온도 범위는 17.2도입니다."가 출력됩니다.
print(f"온도 범위는 {temp_range:.1f}도입니다.")
```

> **참고**
>
> 강의 자료 「리스트에서 최댓값·최솟값·평균 구하기」 절의 "직접 해보기" 상자에 있는 f-string을 다시 보세요. 중괄호 안에서 변수 이름 바로 뒤에 붙인 서식 기호가 "소수 첫째 자리까지만 보여 달라"는 뜻입니다.
>
> 정답 해설: 빈칸은 `:.1f`. `{temp_range:.1f}`는 값을 소수 첫째 자리까지 반올림해 보여 준다(`f`는 소수 형식이라는 뜻). 변수에 저장된 값 자체는 바뀌지 않고 보이는 모양만 바뀐다 — 5단계 `round(average, 1)`과 같은 목적이다.
>
> 예상 결과: `온도 범위는 17.2도입니다.`

## 6단계. 딕셔너리 만들기
**설명 포인트**: 리스트와 곧바로 비교 — "순서 대신 이름표(키)로 값을 구분한다".

**진행 방법**: 사람 딕셔너리(`person`) 두 셀로 감을 잡은 뒤 설비 딕셔너리로 넘어간다.

**예상 결과**: `{'이름': '지민', '나이': 15}` → `지민` → `{'설비번호': 'EQ-01', '공정명': '식각', '온도_섭씨': 301.2}`

### 따라 해보기 — 아주 간단한 딕셔너리
반도체 데이터로 들어가기 전에, 사람 한 명의 정보를 담은 가장 단순한 딕셔너리로 감을 잡습니다
(강의 자료 "먼저 아주 간단한 예로 감을 잡자"). 완성된 코드이니 실행만 해 보세요.

```python
# 사람 한 명의 이름과 나이를 딕셔너리로 만들어 person에 저장하고 출력합니다.
# 실행하면 {'이름': '지민', '나이': 15}가 출력됩니다.
# 딕셔너리: 중괄호 { } 안에 키: 값 쌍을 콤마로 나열합니다. 키는 값에 붙이는 이름표입니다.
person = {"이름": "지민", "나이": 15}
print(person)
```

```python
# 이름표(키) "이름"으로 값을 꺼내 출력합니다.
# 실행하면 지민이 출력됩니다.
# 딕셔너리이름[키]: 리스트의 인덱스 대신 키를 대괄호 안에 넣어 값을 꺼냅니다.
print(person["이름"])
```

### 반도체 데이터로 — 설비 1대를 딕셔너리로
같은 방식으로 설비 1대의 정보(설비번호, 공정명, 온도)를 딕셔너리로 표현합니다.

## 오류 대처 방법
- `IndexError` 발생 시: `len()`으로 리스트 길이를 먼저 확인.

## 13단계. AI에게 코드 생성 요청하고 직접 실행해보기
**설명 포인트**: `max(..., key=lambda ...)`는 오늘 배우지 않은 문법이다 — "AI가 준 코드도 이해하고
써야 한다"는 메시지를 전달하기 좋은 타이밍이다.

프롬프트 예시: "설비 정보 딕셔너리가 여러 개 담긴 리스트에서, 온도가 가장 높은 설비 하나만
찾아주는 코드를 만들어줘."

아래는 AI가 생성해줄 수 있는 예시 코드와 실행 결과다.

**주의**: 강의 자료의 예시 코드는 목록 이름을 `sample_list`로 지어 9단계에서 만든 `equipment_list`(공정명 포함)를 덮어쓰지 않습니다. AI가 준 코드가 `equipment_list = [...]`처럼 같은 이름을 쓰면 내 설비 목록이 새 목록으로 바뀌니, `sample_list` 같은 다른 이름으로 바꿔서 실행하세요.

강의 예시는 목록 이름을 `sample_list`로 지어 9단계 `equipment_list`를 덮어쓰지 않는다(1차 작업의 복원 셀은 필요 없어져 제거했다). 학생이 받은 AI 코드가 `equipment_list = [...]`를 쓰면 이름을 바꾸게 하거나, 9단계 셀을 다시 실행하게 한다.

```python
# AI가 만들어 줄 수 있는 코드 예시입니다(강의 자료와 같음). 설비 3대 중 온도가 가장 높은 설비를 찾아 문장으로 출력합니다.
# 실행하면 "가장 온도가 높은 설비는 EQ-02이고, 온도는 308.7도입니다."가 출력됩니다.
# max(리스트, key=lambda eq: eq["온도_섭씨"]): 각 딕셔너리의 "온도_섭씨" 값을 기준으로 가장 큰 항목을 고릅니다(오늘 배우지 않은 문법).
# 목록 이름을 sample_list로 지어, 9단계에서 만든 equipment_list(공정명 포함)를 덮어쓰지 않습니다.
sample_list = [
    {"설비번호": "EQ-01", "온도_섭씨": 301.2},
    {"설비번호": "EQ-02", "온도_섭씨": 308.7},
    {"설비번호": "EQ-03", "온도_섭씨": 297.5},
]

hottest = max(sample_list, key=lambda eq: eq["온도_섭씨"])
print(f"가장 온도가 높은 설비는 {hottest['설비번호']}이고, 온도는 {hottest['온도_섭씨']}도입니다.")
```

## 14단계. 연습문제 — 압력 리스트와 새 설비 딕셔너리
**설명 포인트**: 7단계(1차시 온도·설비 문장 만들기)와 같은 목적의 단계다. 오늘 배운 리스트
생성·통계 계산(`max`/`min`/`sum`/`len`)과 딕셔너리 키 추가를 새로운 데이터(압력)로 한 번 더
반복 연습시킨다. 빠른 학습자에게는 도전(새 설비 딕셔너리)까지, 느린 학습자에게는 리스트 통계
계산까지만 요구해도 충분하다.

**예상 결과**: `[1010.2, 1005.4, 1022.9, 998.7, 1011.0]` → `1022.9 998.7` → `최댓값: 1022.9, 최솟값: 998.7, 평균: 1009.6`. 도전: `{'설비번호': 'EQ-04', '공정명': '세정', '온도_섭씨': 299.9}` → `{'설비번호': 'EQ-04', '공정명': '세정', '온도_섭씨': 299.9, '압력_Pa': 1009.64}`(강의 연습문제 "정답 보기"와 같은 결과).

```python
# 오늘 측정한 압력 값 5개를 리스트로 만들어 pressures에 저장하고, 잘 저장됐는지 출력해 확인합니다.
# 실행하면 [1010.2, 1005.4, 1022.9, 998.7, 1011.0]이 출력됩니다.
pressures = [1010.2, 1005.4, 1022.9, 998.7, 1011.0]
print(pressures)  # (중간 확인)
```

```python
# 압력의 최댓값과 최솟값을 구해 highest_p, lowest_p에 저장하고, 두 값을 출력해 확인합니다.
# 실행하면 1022.9 998.7이 한 줄에 출력됩니다.
highest_p = max(pressures)
lowest_p = min(pressures)
print(highest_p, lowest_p)  # (중간 확인)
```

```python
# 평균 압력을 계산해 average_p에 저장하고, 최댓값·최솟값·평균을 한 문장으로 출력합니다.
# 실행하면 "최댓값: 1022.9, 최솟값: 998.7, 평균: 1009.6"이 출력됩니다.
average_p = sum(pressures) / len(pressures)

print(f"최댓값: {highest_p}, 최솟값: {lowest_p}, 평균: {round(average_p, 1)}")
```

### 도전. 새 설비 딕셔너리에 압력 추가하기
**시연 방법**: 방금 계산한 `average_p`를 그대로 새 딕셔너리의 값으로 재사용하는 것을 보여주면,
"한 계산 결과를 다른 자료구조에 다시 담을 수 있다"는 감각을 심어주기 좋다.

```python
# 새 설비 EQ-04의 정보를 딕셔너리 eq04로 만들고, 잘 만들어졌는지 출력해 확인합니다.
# 실행하면 {'설비번호': 'EQ-04', '공정명': '세정', '온도_섭씨': 299.9}가 출력됩니다.
# 딕셔너리는 6단계처럼 여러 줄로 적어도 되고, 이렇게 한 줄로 적어도 됩니다.
eq04 = {"설비번호": "EQ-04", "공정명": "세정", "온도_섭씨": 299.9}
print(eq04)  # (중간 확인)
```

```python
# 방금 계산한 평균 압력(average_p)을 "압력_Pa"라는 새 키로 eq04에 추가하고 출력합니다.
# 실행하면 맨 끝에 '압력_Pa': 1009.64가 붙은 딕셔너리가 출력됩니다.
eq04["압력_Pa"] = average_p
print(eq04)
```

## 15단계. AI로 재미있는 미니 프로그램 만들기 🎉 — 오늘의 럭키 설비 뽑기
**설명 포인트**: 8단계(1차시 공정 운세 뽑기)와 같은 목적의 마무리 단계다. 정확한 코드 자체보다
"오늘 만든 리스트·딕셔너리를 AI와 함께 재미있는 프로그램으로 확장해보는 경험"을 즐기도록 유도한다.

프롬프트 예시: "설비 정보 딕셔너리가 여러 개 담긴 리스트(equipment_list)에서, random.choice()로
하나를 무작위로 뽑아서 '오늘의 럭키 설비'라는 제목과 함께 설비번호·공정명·온도를 예쁘게 출력하는
파이썬 코드를 만들어줘."

**주의 안내**: `random` 모듈은 오늘 처음 보는 표현이다 — 리스트·딕셔너리는 오늘 배운 내용이니,
`random.choice()`가 리스트에서 하나를 무작위로 골라준다는 정도만 짚어주면 충분하다.

아래는 AI가 생성해줄 수 있는 예시 코드다(강의 자료 예시와 같은 코드). `import random` 아래의 `random.seed(7)` 줄로 결과를 고정해 두어 몇 번을 실행해도 **EQ-03(포토, 297.5도)**이 뽑힌다(강의 "실행 결과"와 같다). **seed 줄을 지우면 실행할 때마다 뽑히는 설비가 달라진다** — 시연에서 seed 줄을 지우고 여러 번 실행해 무작위를 보여줘도 좋다. 네 번째 설비(EQ-04, 세정, 299.9)는 14단계 `eq04`와 같은 값을 직접 적어 넣은 것이다.

```python
# AI가 만들어 줄 수 있는 "오늘의 럭키 설비 뽑기" 코드 예시입니다(강의 자료와 같음). 설비 4대 중 하나를 무작위로 골라 정보를 출력합니다.
# 실행하면 제목 줄과 뽑힌 설비의 설비번호·공정명·온도가 출력됩니다. random.seed(7) 줄 때문에 몇 번을 실행해도 같은 설비가 뽑히고, 그 줄을 지우면 실행할 때마다 바뀝니다.
# import random: 무작위 기능이 담긴 random 모듈을 불러옵니다. random.choice(리스트)는 리스트에서 하나를 무작위로 고릅니다.
# random.seed(숫자): 무작위 뽑기의 출발점을 정해 두어, 같은 숫자를 쓰면 항상 같은 결과가 나오게 합니다.
import random
random.seed(7)  # 실행할 때마다 같은 결과가 나오도록 고정

equipment_list = [
    {"설비번호": "EQ-01", "공정명": "식각", "온도_섭씨": 301.2},
    {"설비번호": "EQ-02", "공정명": "증착", "온도_섭씨": 308.7},
    {"설비번호": "EQ-03", "공정명": "포토", "온도_섭씨": 297.5},
    {"설비번호": "EQ-04", "공정명": "세정", "온도_섭씨": 299.9},
]

lucky = random.choice(equipment_list)
print("🎉 오늘의 럭키 설비 🎉")
print(f"설비번호: {lucky['설비번호']}")
print(f"공정명: {lucky['공정명']}")
print(f"온도: {lucky['온도_섭씨']}도")
```
