## 1단계. Pandas 불러오기

## 2단계. 데이터 읽기
**설명 포인트**: `read_csv`의 인자가 파일 경로임을 강조. Colab에서는 업로드 후 파일명만 적으면 된다는 점을 시연.

```python
df = pd.read_csv("../../data/weekly/week03/week03_process_filtering.csv")
df.head()
```

**예상 결과**: 8개 열(측정시간, 로트번호, 설비번호, 공정명, 온도_섭씨, 압력_Pa, 가스유량_slm, 합격여부)이 보임. 총 100행 8열.

## 3단계. 열 선택하기
**어려워할 부분**: 여러 열을 선택할 때 대괄호를 한 번만 써서 오류를 겪는 학생이 많음.
`df[['온도_섭씨', '압력_Pa']]`처럼 바깥 대괄호(선택 문법) + 안쪽 대괄호(리스트)가 필요함을 강조.

```python
df["온도_섭씨"]

df[["온도_섭씨", "압력_Pa"]]
```

**예상 결과**: 첫 번째는 Series(1개 열), 두 번째는 DataFrame(2개 열)이 출력됨.

```python
df.dtypes
```

**예상 결과**: `온도_섭씨`, `압력_Pa`, `가스유량_slm`은 `float64`, `합격여부`는 `int64`,
나머지 문자열 열은 `object`.

## 4단계. 조건 검색 — 특정 설비 찾기
**칠판 포인트**: `=`(저장)과 `==`(비교)를 나란히 적어 구분해서 설명.

```python
eq02 = df[df["설비번호"] == "EQ-02"]
print(eq02.shape)
```

**예상 결과**: `(32, 8)` — EQ-02 공정 기록은 32건.

## 5단계. 불합격 행 찾기
**설명 포인트**: 4단계에서 만든 `eq02` 위에 조건을 한 번 더 적용하는 것을 보여준다
(조건 검색을 이어서 적용할 수 있다는 감각).

```python
fail_eq02 = eq02[eq02["합격여부"] == -1]
fail_eq02[["측정시간", "로트번호", "공정명", "합격여부"]]
```

**예상 결과**: 8건이 출력됨. 8 ÷ 32 × 100 = 25.0%.
전체 데이터에서 EQ-02가 아닌 나머지 설비의 불합격 비율(5.9%)과 비교해 설명한다.

## 6단계. 온도 기준으로 정렬하기
**어려워할 부분**: `ascending=False`를 빠뜨리면 오름차순으로 정렬된다는 것을 모를 수 있음.

```python
df.sort_values("온도_섭씨", ascending=False).head(5)
```

**예상 결과**: 상위 5건의 로트가 LOT-0041(EQ-04), LOT-0362(EQ-02), LOT-0338(EQ-03),
LOT-0340(EQ-02), LOT-0399(EQ-03) 순으로 나오며, 모두 `합격여부`가 1(합격)이다.
"온도가 높다고 항상 불합격은 아니다"는 점을 짚어준다(8주차 상관관계 논의의 씨앗).

## 7단계. 확장 실습 — 나머지 설비와 비교
**실습 시간 조절**: 시간이 부족하면 이 단계는 숙제로 돌려도 된다.

```python
others = df[df["설비번호"] != "EQ-02"]
fail_rate_eq02 = (eq02["합격여부"] == -1).mean() * 100
fail_rate_others = (others["합격여부"] == -1).mean() * 100
print(f"EQ-02 불합격 비율: {fail_rate_eq02:.1f}%")
print(f"나머지 설비 불합격 비율: {fail_rate_others:.1f}%")
```

**예상 결과**: `EQ-02 불합격 비율: 25.0%`, `나머지 설비 불합격 비율: 5.9%`.

## 8단계. 오류 대처 방법
- `KeyError` 발생 시: `df.columns`로 정확한 열 이름 확인.
- `SyntaxError`(조건식에 `=` 사용) 발생 시: `==`로 고치도록 안내.
- 필터 결과가 빈 표로 나오면: 문자열 대소문자·띄어쓰기(`"EQ-02"`)가 정확한지 확인.

## 확장 실습(빠른 학습자용)
`&`로 조건 두 개를 동시에 적용해보게 하면 4주차(결측값 필터링) 예고가 된다.

```python
df[(df["설비번호"] == "EQ-02") & (df["합격여부"] == -1)]
```
