학생용 Notebook과 동일한 흐름이며, 예상 결과와 설명 포인트를 함께 담았습니다.

## 1단계. 조건문으로 온도 하나 판정하기
**설명 포인트**: 들여쓰기 4칸이 필수임을 강조.

```python
temperature = 312.6

if temperature > 305:
    print("온도가 너무 높습니다!")
```

## 2단계. if/elif/else로 등급 매기기

```python
score = 75

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "D"

print(f"점수 {score}는 {grade}등급입니다.")
```

## 3단계. for로 리스트 순회하기
**어려워할 부분**: for 다음의 변수명이 자유롭다는 것을 x로 바꿔 실행해서 보여주면 효과적.

```python
temperatures = [295.4, 298.1, 312.6, 301.0, 296.8]

for temp in temperatures:
    print(temp)
```

## 4단계. for + if로 정상 개수 세기
**설명 포인트**: normal_count = 0을 for 안에 넣으면 안 되는 이유를 실제로 넣어 실행해 보여준다.

```python
normal_count = 0
for temp in temperatures:
    if 295 <= temp <= 305:
        normal_count += 1

print(f"정상 범위인 값은 {normal_count}개입니다.")
```

## 5단계. 딕셔너리의 리스트에서 조건에 맞는 것 골라내기

```python
equipment_list = [
    {"설비번호": "EQ-01", "공정명": "식각", "온도_섭씨": 301.2},
    {"설비번호": "EQ-02", "공정명": "증착", "온도_섭씨": 308.7},
    {"설비번호": "EQ-03", "공정명": "포토", "온도_섭씨": 297.5},
]

hot_equipment = []
for eq in equipment_list:
    if eq["온도_섭씨"] > 305:
        hot_equipment.append(eq["설비번호"])

print(hot_equipment)
```

## 6단계. 함수 만들기 — 평균 계산 함수
**설명 포인트**: def(정의)와 호출은 다른 단계임을 강조 — 정의만 실행해도 출력이 없다는 것을 보여준다.

```python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)

print(calculate_average(temperatures))
```

## 7단계. 함수 만들기 — 합격 판정 함수

```python
def judge_pass(temp):
    if 295 <= temp <= 305:
        return "합격"
    else:
        return "불합격"

for temp in temperatures:
    print(f"{temp}도 → {judge_pass(temp)}")
```

## 8단계(종합). 함수 + 반복문으로 여러 로트 판정하기
**시연 방법**: 학생에게 lots 리스트에 로트를 하나 더 추가해보라고 하면 참여도가 높아진다.

```python
lots = [
    {"로트번호": "LOT-0101", "온도_섭씨": 299.5},
    {"로트번호": "LOT-0102", "온도_섭씨": 306.2},
    {"로트번호": "LOT-0103", "온도_섭씨": 301.0},
]

fail_count = 0
for lot in lots:
    result = judge_pass(lot["온도_섭씨"])
    print(f"{lot['로트번호']}: {result}")
    if result == "불합격":
        fail_count += 1

print(f"불합격 로트 수: {fail_count}개")
```

## 9단계. 오늘의 학습을 한 문장으로 정리하기
**진행 방법**: 조건문·반복문·함수 세 가지를 각각 한 마디씩 붙여 말하게 하면 문장이 쉽게 만들어진다. "~할 때 쓴다" 형태로 유도한다.
**설명 포인트**: 결과 해석 문장 쓰기는 1차시부터 10차시까지 매 차시 빠지지 않는 고정 활동이다(docs/curriculum.md 「차시 간 연결 원칙」). 코드를 친 것으로 끝내지 않고 자기 말로 바꿔 말해보게 하는 것이 이 과정의 목표 — "데이터로 공정 상태를 설명할 수 있는 사람" — 에 직접 닿는 활동이므로 시간이 모자라도 2~3분은 반드시 확보한다.

> (예시 답안) 조건문은 상황에 따라 다르게 판단하고, 반복문은 여러 값을 하나씩 순회하며, 함수는 반복해서 쓰는 코드를 이름 붙여 재사용한다.

## 오류 대처 방법

- `SyntaxError` 발생 시: 조건문에서 `=` 대신 `==`를 썼는지 확인.

## 확장 실습(빠른 학습자용)
매개변수 2개(value, limit)를 받는 범용 판정 함수를 만들어 압력에도 재사용해보게 한다.

```python
def judge_range(value, low, high):
    return "정상" if low <= value <= high else "비정상"

print(judge_range(1030.0, 995, 1025))
```

## 10단계. AI에게 질문하며 더 알아보기
**설명 포인트**: 리스트 컴프리헨션 질문이 나오면, 오늘 배운 for+if 방식이 먼저이고 압축 표현은
"이해한 뒤에 선택적으로" 쓰는 것임을 강조한다.

- "리스트 컴프리헨션(list comprehension)이 뭔가요?"
- "재귀 함수(recursive function)가 뭔가요?"

## 11단계. AI에게 코드 생성 요청하고 직접 실행해보기
**설명 포인트**: 함수가 딕셔너리를 반환하는 패턴은 오늘 배운 return 개념 그대로다 — AI 코드가
"새로운 문법"이 아니라 "오늘 배운 것의 조합"임을 보여주기 좋은 예시다.

프롬프트 예시: "로트번호와 온도가 담긴 딕셔너리 리스트를 받아서, 합격/불합격 개수를 각각 세어
딕셔너리로 반환하는 함수를 만들어줘."

아래는 AI가 생성해줄 수 있는 예시 코드와 실행 결과다.

```python
def count_results(lots):
    counts = {"합격": 0, "불합격": 0}
    for lot in lots:
        result = judge_pass(lot["온도_섭씨"])
        counts[result] += 1
    return counts

print(count_results(lots))
```

## 12단계. 연습문제 — 온도 등급 나누기
**설명 포인트**: if/elif로 등급을 나누는 판정 로직과 딕셔너리로 결과를 집계하는 방식을 결합하는
연습이다 — 앞서 7~8단계에서 만든 judge_pass + fail_count 패턴을 세 갈래 등급으로 확장한 것임을
짚어준다.

```python
def classify_temp(temp):
    if temp > 305:
        return "위험"
    elif 295 <= temp <= 305:
        return "정상"
    else:
        return "주의"

counts = {"위험": 0, "정상": 0, "주의": 0}
for temp in temperatures:
    grade = classify_temp(temp)
    counts[grade] += 1

print(counts)
```

## 13단계. AI로 재미있는 미니 프로그램 만들기 🎉 — 온도 뽑기 심판 프로그램
**설명 포인트**: `judge_pass` 함수를 다시 정의하지 않고 그대로 재사용하는 것이 핵심이다 — 오늘
배운 함수가 다른 맥락(무작위 온도 뽑기)에서도 그대로 재사용될 수 있음을 보여주는 예시로
활용한다.

프롬프트 예시: "0부터 320 사이의 임의의 온도를 5번 뽑아서, judge_pass 같은 함수로 합격/불합격을
판정하고 합격이면 ✅, 불합격이면 ❌ 이모지와 함께 심판처럼 발표하는 파이썬 코드를 만들어줘.
random 모듈을 사용해도 좋아."

아래는 AI가 생성해줄 수 있는 예시 코드와 실행 결과다(앞서 만든 `judge_pass` 함수를 그대로
재사용한다).

```python
import random

random.seed(3)

print("🎤 오늘의 온도 뽑기 심판 프로그램 🎤")
for i in range(1, 6):
    temp = round(random.uniform(290, 320), 1)
    result = judge_pass(temp)
    mark = "✅" if result == "합격" else "❌"
    print(f"{i}번째 로트: {temp}도 → {result} {mark}")
```
