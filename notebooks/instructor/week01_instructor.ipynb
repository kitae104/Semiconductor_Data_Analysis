## 7단계. 변수 만들고 값 확인하기
**설명 포인트**: `=`는 수학의 '같다'가 아니라 대입 연산자임을 강조. 칠판에 상자 그림을 그려 '오른쪽 값이 왼쪽 이름의 상자에 들어간다'고 설명하면 효과적.

````python
pass_count = 34
fail_count = 6
print(pass_count)
print(fail_count)
````

## 8단계. f-string으로 문장 만들기
**어려워할 부분**: 문자열 앞의 `f`를 빼먹으면 중괄호가 그대로 출력된다는 점을 `print("합격 {pass_count}건")`과 비교해 직접 보여주면 효과적.

````python
print(f"합격 {pass_count}건, 불합격 {fail_count}건입니다.")
````

## 9단계. input()으로 값 입력받기
**시연 방법**: 실제 수업에서는 아래 코드를 그대로 실행해 학생이 입력하는 과정을 보여준다.
```python
name = input("이름을 입력하세요: ")
```
이 노트북은 `validate_notebooks.py`로 자동 실행되므로, `input()`이 멈추지 않도록 아래 셀에서는 예시 값을 직접 대입한다.

````python
# 자동 실행 검증에서는 input()이 멈추지 않도록 예시 값을 대입합니다.
# 실제 수업에서는 name = input("이름을 입력하세요: ")으로 대체하세요.
name = "지민"
print(f"안녕하세요, {name}님!")
````

## 10단계. 나만의 자기소개 만들기
**시연 방법**: 학생 한 명을 지목해 실시간으로 입력받아 결과를 함께 확인하면 좋다.
