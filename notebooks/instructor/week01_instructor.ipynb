## 4단계. 합격/불합격 개수
**어려워할 부분**: `value_counts()`가 자동으로 개수 많은 순으로 정렬된다는 점을 모를 수 있음.

```python
df["합격여부"].value_counts()
```

**예상 결과**: `1: 34건, -1: 6건` (실행 환경에 따라 약간 다를 수 있음 — 시드가 고정되어 있으므로 기본적으로 동일).
