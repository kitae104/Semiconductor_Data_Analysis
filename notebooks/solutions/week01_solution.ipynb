## 4단계. print()와 f-string으로 문장 만들기

f-string은 따옴표 앞에 `f`를 붙이고, 값이 들어갈 자리에 `{변수 이름}`을 씁니다. **주의**: `f`를 빼먹으면 중괄호가 글자 그대로 나옵니다.

```python
print("메뉴:", menu, "잔 수:", cups)
```

## 5단계. 계산한 값을 문장에 넣기

따뜻한 커피 6잔, 아이스 4잔이 팔렸습니다.

```python
hot_count = 6
ice_count = 4

total = hot_count + ice_count
ice_rate = ice_count / total * 100
print(total, ice_rate)
```

```python
print(f"전체 {total}잔 중 아이스는 {ice_rate}%입니다.")
```

## 6단계. 반도체 데이터로 확인하기 — 공장 기록 40건

공장 기록표(교육용 가상 데이터) 40건에서 `합격여부`를 센 결과입니다. 1 = 합격, -1 = 불합격. Orange의 Distributions 막대 높이와 같은 숫자입니다.

```python
pass_count = 32
fail_count = 8
total_count = pass_count + fail_count

print(f"오늘 기록 {total_count}건 중 합격은 {pass_count}건, 불합격은 {fail_count}건입니다.")
```

```python
fail_rate = fail_count / total_count * 100
print(f"불합격률은 {fail_rate}%입니다.")
```
