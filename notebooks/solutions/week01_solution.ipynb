# 1차시 실습 — 변수와 출력 (정답)

강의에서 데이터 표를 눈으로 정리한 결과(합격 32건, 불합격 8건)로 시작합니다.

## 1단계. 변수 만들고 값 확인하기

```python
pass_count = 32
fail_count = 8

print(pass_count)
print(fail_count)
```

### 자료형 확인하기

```python
temperature = 300.1        # 실수(float)
equipment = "EQ-03"        # 문자열(str)
is_pass = True             # 불리언(bool)

print(type(pass_count))
print(type(temperature))
print(type(equipment))
print(type(is_pass))
```

## 2단계. print()와 f-string으로 문장 만들기

```python
print("합격:", pass_count, "불합격:", fail_count)
```

### f-string 기본

```python
print(f"오늘 합격은 {pass_count}건, 불합격은 {fail_count}건입니다.")
```

### 계산한 값 넣기

```python
total = pass_count + fail_count
fail_rate = fail_count / total * 100

print(f"전체 {total}건 중 불합격률은 {fail_rate:.1f}%입니다.")
```

### 직접 해보기 — 설비 온도 문장

```python
print(f"{equipment} 설비의 온도는 {temperature}도입니다.")
```

## 3단계. input()으로 값 입력받기
자동 실행을 위해 `input()` 대신 예시 입력값을 넣었습니다.

```python
name = "지민"  # 실제로는 input("이름을 입력하세요: ")
print(f"안녕하세요, {name}님! 반갑습니다.")
```

### 입력값을 숫자로 바꾸기

```python
age_text = "24"  # 실제로는 input("나이를 입력하세요: ")
print(type(age_text))

age = int(age_text)
print(f"내년이면 {age + 1}살이 되네요!")

# print(age_text + 1)  # 주석을 풀면 TypeError: 문자열과 숫자는 더할 수 없다
```

## 4단계. 나만의 자기소개 만들기

```python
my_name = "하은"  # 실제로는 input("이름을 입력하세요: ")
favorite_process = "식각"  # 실제로는 input("가장 기억에 남는 공정을 입력하세요: ")

print(f"안녕하세요! 저는 {my_name}입니다.")
print(f"오늘 배운 공정 중에서는 '{favorite_process}' 공정이 가장 기억에 남아요.")
```

### 도전 문제 — 합격률 문장 만들기 (빠른 학습자용)

```python
total = pass_count + fail_count
pass_rate = pass_count / total * 100
print(total, pass_rate)
```

```python
print(f"합격률은 {pass_rate:.1f}%입니다.")
```

## 5단계. AI에게 질문하며 더 알아보기
질문 예시는 강의 "AI로 한 걸음 더" 절에 있습니다. 답은 학생마다 다릅니다.

## 6단계. AI에게 코드 생성 요청하고 직접 실행해보기
강의의 "AI가 만들어줄 수 있는 코드 예시"입니다.

```python
pass_count = 32
fail_count = 8
total = pass_count + fail_count
pass_rate = pass_count / total * 100

if pass_rate >= 90:
    print(f"합격률 {pass_rate:.1f}% → 정상")
else:
    print(f"합격률 {pass_rate:.1f}% → 점검 필요")
```

## 7단계. 연습문제 — 클린룸 온도로 문장 만들기

```python
temp_eq02 = 23.5
temp_eq05 = 24.1
```

```python
avg_temp = (temp_eq02 + temp_eq05) / 2
print(f"두 설비의 평균 온도는 {avg_temp:.1f}도입니다.")
```

## 8단계. AI로 재미있는 미니 프로그램 만들기
강의 예시 코드입니다. `random.seed(42)`로 결과를 고정했습니다.

```python
import random

random.seed(42)  # 실행할 때마다 같은 결과가 나오도록 고정

name = "지민"  # 실제로는 input("이름을 입력하세요: ")
fortunes = [
    "오늘은 식각 공정에서 대박이 납니다!",
    "포토 공정에서 살짝 조심하세요.",
    "증착 공정 결과가 아주 좋을 예감입니다.",
    "세정을 꼼꼼히 하면 하루 종일 순조롭습니다.",
    "산화 공정에서 뜻밖의 좋은 소식이 있을 거예요.",
]

today_fortune = random.choice(fortunes)
print(f"{name}님의 오늘의 반도체 공정 운세: {today_fortune}")
```
