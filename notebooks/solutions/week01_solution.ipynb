## 1단계. 변수 만들고 값 확인하기

```python
# 합격 개수(32)와 불합격 개수(8)를 변수에 저장한 뒤, print()로 다시 꺼내 출력합니다.
# 실행하면 32와 8이 한 줄에 하나씩 출력됩니다.
# print(): 괄호 안의 값을 화면에 보여주는 함수입니다.
# 값에 이름을 붙여 저장한다(대입).
pass_count = 32
fail_count = 8

# 이름(변수)만 적으면 그 안에 든 값을 다시 꺼내 쓸 수 있다.
print(pass_count)
print(fail_count)
```

### 따라 해보기 — 자료형 확인하기
변수에 담긴 값의 종류(자료형)를 `type()`으로 확인합니다.

```python
# 실수·문자열·불리언 값을 변수 세 개에 저장하고, type()으로 각각의 자료형을 확인합니다.
# 실행하면 <class 'float'>, <class 'str'>, <class 'bool'>이 차례로 출력됩니다.
# type(): 괄호 안 값의 자료형(종류)을 알려주는 함수입니다.
temperature = 300.1        # 실수(float)
equipment = "EQ-03"        # 문자열(str) — 따옴표로 감싼다
is_pass = True              # 불리언(bool) — True/False만 가능

print(type(temperature))
print(type(equipment))
print(type(is_pass))
```

## 2단계. f-string으로 문장 만들기

### 따라 해보기 — print()로 여러 값 한 번에 출력하기

```python
# print() 안에 콤마(,)로 글자와 변수를 나열해 한 줄에 출력합니다.
# 실행하면 "합격: 32 불합격: 8"이 출력됩니다 — 값 사이에 공백이 하나씩 들어간 것을 확인해 보세요.
print("합격:", pass_count, "불합격:", fail_count)
```

### f-string 기본 — 문장 속에 값 끼워 넣기

```python
# f-string으로 문장 속 원하는 위치에 변수 값을 끼워 넣어 출력합니다.
# 실행하면 "오늘 합격은 32건, 불합격은 8건입니다."가 출력됩니다.
# f-string: 문자열 앞에 f를 붙이고, 값이 들어갈 자리에 {변수}를 적습니다.
print(f"오늘 합격은 {pass_count}건, 불합격은 {fail_count}건입니다.")
```

### 따라 해보기 — f-string 안에서 계산하기

```python
# 전체 건수(total)와 불합격률(fail_rate)을 계산한 뒤, 불합격률을 소수 첫째 자리까지 출력합니다.
# 실행하면 "전체 40건 중 불합격률은 20.0%입니다."가 출력됩니다.
# {fail_rate:.1f}: 콜론(:) 뒤의 .1f는 "소수 첫째 자리까지 보여 달라"는 서식입니다.
total = pass_count + fail_count
fail_rate = fail_count / total * 100

print(f"전체 {total}건 중 불합격률은 {fail_rate:.1f}%입니다.")
```

### 직접 해보기 — 설비 이름·온도로 나만의 f-string 문장 만들기
강의 "직접 해보기" 상자의 예시 문장입니다. 1단계에서 만든 설비 이름·온도 변수를 다시 씁니다.

```python
# 1단계에서 만든 설비 이름 변수와 온도 변수를 f-string 문장에 끼워 넣어 출력합니다.
# 실행하면 "EQ-03 설비의 온도는 300.1도입니다."가 출력됩니다.
print(f"{equipment} 설비의 온도는 {temperature}도입니다.")
```

> **참고**
> - 따옴표 앞 빈칸: 강의 자료 "f-string으로 문장 속에 값 끼워 넣기" 절의 "사용법" 줄을 다시 보세요.
>   따옴표 앞에 글자 하나를 붙여야 중괄호 `{ }` 안의 변수 이름이 그 값으로 바뀝니다. 빼먹으면 중괄호가 글자 그대로 출력됩니다
>   ("주의할 점 & 자주 발생하는 오류" 절의 "주의 — f-string 흔한 실수").
> - 중괄호 안 빈칸: 강의 자료 "자료형 — 정수·실수(숫자형), 문자열형, 불리언" 절의 "자료형 확인하기" 코드에서
>   300.1을 저장한 변수 이름을 찾아보세요. 중괄호 안에 변수 이름을 적으면 그 변수에 든 값이 문장 속에 끼워집니다.
>
> **정답 해설:** 따옴표 앞에 `f`, 중괄호 안에 `temperature`를 채운다. `f`가 있어야 `{equipment}`·`{temperature}`가
> 각각 `EQ-03`, `300.1`로 바뀐다.

> 참고: 자동 실행 환경에서는 input()이 멈추지 않도록 예시 값을 대신 사용합니다. 직접 실행할 때는 `name = input("이름을 입력하세요: ")`을 사용하세요.

```python
# input()으로 이름을 입력받아 name에 저장하고, f-string으로 인사 문장을 출력합니다.
# 실행하면 "안녕하세요, 지민님! 반갑습니다."처럼 입력한 이름이 들어간 문장이 출력됩니다.
# input("질문"): 화면에 질문을 보여주고, 키보드로 입력한 내용을 결과값으로 돌려줍니다.
name = "지민"  # 실제로는 input("이름을 입력하세요: ")
print(f"안녕하세요, {name}님! 반갑습니다.")
```

### 따라 해보기 — 입력값을 숫자로 바꾸기
`input()`의 결과는 항상 문자열(str)이므로, 계산하려면 `int()`로 바꿉니다(아래는 입력값 24를 예시 값으로 대입).

```python
# 나이를 입력받아 age_text에 저장하고, 그 자료형을 확인합니다.
# 실행하면 <class 'str'>이 출력됩니다 — 숫자를 입력해도 input()의 결과는 문자열입니다.
age_text = "24"  # 실제로는 input("나이를 입력하세요: ") — age_text는 항상 문자열!
print(type(age_text))  # (중간 확인)
```

```python
# int()로 문자열 age_text를 정수 age로 바꾼 뒤, 1을 더해 내년 나이를 출력합니다.
# 실행하면 "내년이면 25살이 되네요!"가 출력됩니다(입력값 24 기준).
# int(): 괄호 안의 값을 정수로 바꾸는 함수입니다. 소수점이 있는 숫자는 float()를 씁니다.
age = int(age_text)                        # 문자열 → 정수로 변환

print(f"내년이면 {age + 1}살이 되네요!")

# 아래 줄 맨 앞의 #을 지우고 실행하면 TypeError가 납니다(문자열과 숫자는 더할 수 없습니다).
# print(age_text + 1)
```

### 도전 문제 1 — 합격률 계산해 출력하기 (빠른 학습자용)
강의 자료 "도전 실습" 절 도전 문제의 예시 답안입니다(계산 / 출력 두 셀).

```python
# 전체 건수(total)와 합격률(pass_rate)을 계산해 변수에 저장하고, 두 값을 출력해 확인합니다.
# 실행하면 40 80.0이 출력됩니다 — 합격률은 100을 곱한 백분율(%) 숫자입니다.
total = pass_count + fail_count
pass_rate = pass_count / total * 100
print(total, pass_rate)  # (중간 확인)
```

> **참고**
> - 첫 번째 빈칸(전체 건수): 강의 자료 "도전 실습" 절의 첫 줄, 또는 "f-string으로 문장 속에 값 끼워 넣기" 절의
>   "f-string 안에서 계산하기" 코드 첫 줄을 다시 보세요. 두 변수 사이에 계산 기호 하나를 넣으면 계산한 결과가 왼쪽 이름에 저장됩니다.
> - 두 번째 빈칸(합격률): 강의 자료 "도전 실습" 절의 둘째 줄을 보세요. 2단계에서 불합격률을 `fail_count / total * 100`으로
>   구한 것과 같은 모양이고, 나누어지는 앞자리만 1단계에서 합격 개수를 저장한 변수로 바뀝니다.
>
> **정답 해설:** 첫 빈칸은 `+`(합격 32 + 불합격 8 = 전체 40), 둘째 빈칸은 `pass_count`(32 ÷ 40 × 100 = 80.0).

```python
# 계산한 합격률을 소수 첫째 자리까지 보여주는 f-string 문장으로 출력합니다.
# 실행하면 "합격률은 80.0%입니다."가 출력됩니다.
print(f"합격률은 {pass_rate:.1f}%입니다.")
```

> **참고**
> - 서식 빈칸: 강의 자료 "f-string으로 문장 속에 값 끼워 넣기" 절의 "소수점 자리수 지정" 줄(또는 "도전 실습" 절 셋째 줄)을
>   다시 보세요. 중괄호 안에서 변수 이름 뒤에 콜론을 쓰고 서식을 붙이면, 값을 소수 몇째 자리까지 보여줄지 정할 수 있습니다.
>
> **정답 해설:** `.1f`를 채운다(소수 첫째 자리까지). `:.2f`로 바꾸면 `80.00`처럼 소수 둘째 자리까지 보여 준다.

## 6단계. AI에게 코드 생성 요청하고 직접 실행해보기
프롬프트 예시: "pass_count, fail_count 변수로 합격률을 계산해서 90% 이상이면 '정상', 아니면
'점검 필요'라고 출력하는 코드를 만들어줘."

아래는 AI가 생성해줄 수 있는 예시 코드와 실행 결과입니다(강의 자료 예시와 같게 맨 위에서
`pass_count`, `fail_count`를 다시 저장하므로 이 셀만 따로 실행해도 동작합니다).

```python
# AI가 만들어 줄 수 있는 코드 예시입니다. 합격률을 계산해 90% 이상이면 "정상", 아니면 "점검 필요"를 출력합니다.
# 실행하면 "합격률 80.0% → 점검 필요"가 출력됩니다.
# if/else(조건문): 조건이 맞으면 if 아래 줄을, 아니면 else 아래 줄을 실행합니다(3차시에서 자세히 배웁니다).
pass_count = 32
fail_count = 8
total = pass_count + fail_count
pass_rate = pass_count / total * 100

if pass_rate >= 90:
    print(f"합격률 {pass_rate:.1f}% → 정상")
else:
    print(f"합격률 {pass_rate:.1f}% → 점검 필요")
```

## 7단계. 연습문제 — 온도·설비 이름으로 문장 만들기

```python
# 두 설비의 온도를 각각 변수에 저장하고, 저장한 값을 문장으로 출력합니다.
# 실행하면 "EQ-02 온도: 23.5도, EQ-05 온도: 24.1도"가 출력됩니다.
temp_eq02 = 23.5
temp_eq05 = 24.1
print(f"EQ-02 온도: {temp_eq02}도, EQ-05 온도: {temp_eq05}도")
```

```python
# 두 온도의 평균을 계산해 avg_temp에 저장하고, 소수 첫째 자리까지(:.1f) 출력합니다.
# 실행하면 "두 설비의 평균 온도는 23.8도입니다."가 출력됩니다.
avg_temp = (temp_eq02 + temp_eq05) / 2
print(f"두 설비의 평균 온도는 {avg_temp:.1f}도입니다.")
```

### 도전. 나만의 문장 만들기
습도까지 변수로 저장하고 온도·습도를 모두 포함한 문장을 출력한 예시 답안입니다(문장 표현은 달라도 됩니다).

```python
# 두 설비의 습도를 각각 변수에 저장하고, 잘 저장됐는지 값을 출력해 확인합니다.
# 실행하면 45.2 44.8이 한 줄에 출력됩니다.
humidity_eq02 = 45.2
humidity_eq05 = 44.8
print(humidity_eq02, humidity_eq05)  # (중간 확인)
```

```python
# 온도와 습도를 함께 담은 나만의 문장을 f-string으로 출력합니다.
# 실행하면 설비별로 온도와 습도가 들어간 문장이 출력됩니다(문장 표현은 사람마다 달라도 됩니다).
print(f"EQ-02: 온도 {temp_eq02:.1f}도 · 습도 {humidity_eq02:.1f}%")
print(f"EQ-05: 온도 {temp_eq05:.1f}도 · 습도 {humidity_eq05:.1f}%")
```

## 8단계. AI로 재미있는 미니 프로그램 만들기 🎉
프롬프트 예시: "input()으로 이름을 입력받고, 반도체 공정 운세 문장 리스트 중 하나를 무작위로
골라 이름과 함께 출력하는 파이썬 코드를 만들어줘."

아래 예시 코드는 `random.seed(42)`로 무작위 결과를 고정해 몇 번을 실행해도 같은 운세가 나옵니다.
seed 줄을 지우면 실행할 때마다 운세가 달라집니다.

```python
# AI가 만들어 줄 수 있는 "반도체 공정 운세 뽑기" 코드 예시입니다. 운세 문장 5개 중 하나를 무작위로 골라 이름과 함께 출력합니다.
# 실행하면 "지민님의 오늘의 반도체 공정 운세: 오늘은 식각 공정에서 대박이 납니다!"가 출력됩니다(seed 42로 고정).
# import random: 무작위 기능이 담긴 random 모듈을 불러옵니다. random.choice(리스트)는 리스트에서 하나를 무작위로 고릅니다.
# random.seed(숫자): 무작위 뽑기의 출발점을 정합니다. 같은 숫자를 쓰면 몇 번을 실행해도 같은 결과가 나옵니다.
import random

random.seed(42)  # 실행할 때마다 같은 결과가 나오도록 고정

name = "지민"  # 실제로는 input("이름을 입력하세요: ")
fortunes = [
    "오늘은 식각 공정에서 대박이 납니다!",
    "포토 공정에서 살짝 조심하세요.",
    "증착 공정 결과가 아주 좋을 예감입니다.",
    "세정을 꼼꼼히 하면 하루 종일 순조롭습니다.",
    "산화 공정에서 뜻밖의 좋은 소식이 있을 거예요.",
]

today_fortune = random.choice(fortunes)
print(f"{name}님의 오늘의 반도체 공정 운세: {today_fortune}")
```
