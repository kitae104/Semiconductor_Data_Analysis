## 1단계. 변수 만들고 값 확인하기

```python
coffee_price = 4500
cups = 3

print(coffee_price)
print(cups)
```

## 2단계. 자료형 확인하기

```python
rating = 4.5               # 실수(float)
menu = "아메리카노"        # 문자열(str)
is_hot = True              # 불리언(bool)

print(type(cups))
print(type(rating))
print(type(menu))
print(type(is_hot))
```

## 3단계. print()와 f-string으로 문장 만들기

```python
print("메뉴:", menu, "잔 수:", cups)
```

```python
print(f"{menu} {cups}잔 주문이 들어왔습니다.")
```

## 4단계. 계산한 값을 문장에 넣기

```python
hot_count = 7
ice_count = 5

total = hot_count + ice_count
ice_rate = ice_count / total * 100
print(total, ice_rate)
```

```python
print(f"전체 {total}잔 중 아이스 비율은 {ice_rate:.1f}%입니다.")
```

## 7단계. 반도체 데이터 실습 — 비율 계산

### 문제 8 — 불합격률

### 문제 9 — 세정 공정의 불합격률

## 9단계. AI에게 질문하며 더 알아보기

## 10단계. AI에게 코드 생성 요청하고 직접 실행해보기

```python
pass_count = 32
fail_count = 8
total = pass_count + fail_count
pass_rate = pass_count / total * 100

if pass_rate >= 90:
    print(f"합격률 {pass_rate:.1f}% → 정상")
else:
    print(f"합격률 {pass_rate:.1f}% → 점검 필요")
```
