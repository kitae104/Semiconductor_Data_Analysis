# 2차시 실습 — 리스트와 딕셔너리 (정답)

오늘 값은 교육용 가상 데이터입니다. 정상 온도 범위 295~305도는 학습용 기준입니다.

## 1단계. 리스트 만들기

```python
fruits = ["사과", "바나나", "포도"]
print(fruits)
```

```python
temperatures = [295.4, 298.1, 312.6, 301.0, 296.8]
print(temperatures)
```

## 2단계. 인덱스로 값 하나 꺼내기

```python
print(temperatures[0])    # 첫 번째 값
print(temperatures[2])    # 세 번째 값
print(temperatures[-1])   # 마지막 값
```

```python
print(temperatures[1])    # 두 번째 값(첫 번째가 아니다!)
# print(temperatures[5])  # 칸은 0~4번뿐 → IndexError
```

## 3단계. 슬라이싱으로 여러 개 꺼내기

```python
print(temperatures[1:3])   # 1번 칸부터 3번 칸 앞까지
print(temperatures[:2])    # 처음부터 2번 칸 앞까지
print(temperatures[2:])    # 2번 칸부터 끝까지
```

### 직접 해보기 — 뒤에서 두 개 꺼내기

```python
print(temperatures[-2:])
```

## 4단계. 개수 세기와 값 추가하기

```python
print(len(temperatures))

temperatures.append(303.2)
print(temperatures)
print(len(temperatures))
```

### 빈 리스트에서 시작해 채우기

```python
new_temps = []
new_temps.append(299.5)
new_temps.append(300.2)
print(new_temps)
```

## 5단계. 최고·최저·평균 구하기

```python
print(max(temperatures))
print(min(temperatures))
print(sum(temperatures))
```

```python
average = sum(temperatures) / len(temperatures)
print(f"평균 온도는 {average:.1f}도입니다.")
```

### 직접 해보기 — 온도 범위(최고 - 최저)

```python
temp_range = max(temperatures) - min(temperatures)
print(f"온도 범위는 {temp_range:.1f}도입니다.")
```

## 11단계. AI에게 코드 생성 요청하고 직접 실행해보기
강의의 "AI가 만들어줄 수 있는 코드 예시"입니다.

```python
temps = [295.4, 298.1, 312.6, 301.0, 296.8]
report = {
    "최고": max(temps),
    "최저": min(temps),
    "평균": round(sum(temps) / len(temps), 1),
}
print(report)
```

## 12단계. 연습문제 — EQ-02 설비 온도 요약하기

```python
eq02_temps = [302.5, 299.8, 298.8, 305.6]
summary = {"설비번호": "EQ-02"}
summary["최고온도"] = max(eq02_temps)
summary["최저온도"] = min(eq02_temps)
print(summary)
```

```python
eq02_avg = sum(eq02_temps) / len(eq02_temps)
print(f"EQ-02 평균 온도는 {eq02_avg:.1f}도입니다.")
```

## 13단계. AI로 재미있는 미니 프로그램 만들기
강의 예시 코드입니다. `random.seed(7)`로 결과를 고정했습니다.

```python
import random
random.seed(7)  # 실행할 때마다 같은 결과가 나오도록 고정

machines = [
    {"설비번호": "EQ-01", "공정명": "식각", "온도_섭씨": 301.2},
    {"설비번호": "EQ-02", "공정명": "증착", "온도_섭씨": 308.7},
    {"설비번호": "EQ-03", "공정명": "포토", "온도_섭씨": 297.5},
    {"설비번호": "EQ-04", "공정명": "세정", "온도_섭씨": 299.9},
]

lucky = random.choice(machines)
print("*** 오늘의 럭키 설비 ***")
print(f"설비번호: {lucky['설비번호']}")
print(f"공정명: {lucky['공정명']}")
print(f"온도: {lucky['온도_섭씨']}도")
```
