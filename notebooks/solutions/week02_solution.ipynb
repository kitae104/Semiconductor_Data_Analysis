## 1단계. 리스트 만들고 인덱스로 꺼내기

리스트는 대괄호 `[ ]` 안에 값을 쉼표로 늘어놓은 것입니다. 칸 번호(인덱스)는 **0부터** 셉니다.

```python
week_temps = [21.5, 23.0, 30.5, 24.0, 22.5, 25.0, 23.5]   # 월~일 낮 기온
print(week_temps)
```

```python
print(week_temps[0])    # 첫 번째 값(월요일)
print(week_temps[2])    # 세 번째 값(수요일)
print(week_temps[-1])   # 마지막 값(일요일)
```

## 2단계. 개수 세고 값 추가하기 — len(), append()

```python
cart = ["우유", "빵", "계란"]
print(len(cart))
```

## 3단계. 최고·최저·평균 구하기 — max(), min(), sum()

```python
print(max(week_temps))
print(min(week_temps))
print(sum(week_temps))
```

평균 = 합계 ÷ 개수입니다. `round(값, 1)`은 소수 첫째 자리까지 반올림합니다.

```python
average = round(sum(week_temps) / len(week_temps), 1)
print(f"이번 주 평균 기온은 {average}도입니다.")
```

## 4단계. 딕셔너리 만들고 키로 꺼내기

딕셔너리는 중괄호 `{ }` 안에 `"이름표": 값`을 쌍으로 적습니다. 꺼낼 때는 대괄호에 이름표(키)를 씁니다.

```python
student = {
    "이름": "지민",
    "나이": 15,
    "반": 3,
}
print(student)
```

## 8단계. AI로 한 걸음 더

ChatGPT에 이렇게 물어봤습니다.

> 파이썬 리스트 eq03_temps = [303.8, 303.4, 292.2, 308.5, 307.5, 298.5, 304.0]에서 295 이상 305 이하 범위를 벗어난 값만 골라 개수와 함께 출력하는 짧은 코드를 만들어 줘.

아래는 받은 코드의 예시입니다. `for`와 `if`는 3차시에 배웁니다. 지금은 실행해서 결과만 확인하세요. 295~305는 이 과정에서 쓰는 연습용 기준입니다.

```python
eq03_temps = [303.8, 303.4, 292.2, 308.5, 307.5, 298.5, 304.0]

out_of_range = []
for t in eq03_temps:
    if t < 295 or t > 305:
        out_of_range.append(t)

print(f"범위 밖 온도: {out_of_range}")
print(f"범위 밖 개수: {len(out_of_range)}개")
```
