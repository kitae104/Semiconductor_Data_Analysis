# 2주차 정답 — 파이썬 기초 Ⅰ — 변수, 자료형, 리스트, 딕셔너리

오늘의 질문: **측정값 하나하나를 어떻게 이름 붙여 기억하고, 여러 개를 한 번에 담을 수 있을까?**

> 🚧 이 노트북은 v2 커리큘럼 구성이 확정된 직후 작성된 **개요(skeleton)**입니다. 다룰 단계와 순서는 정해졌지만, 구체적인 코드·설명·빈칸 문항은 아직 채워지지 않았습니다. 구성이 최종 확정되면 이어서 상세 내용을 작성합니다.

## 다룰 실습 단계(초안)

1. 온도 값 하나를 변수에 저장해보기
2. 하루치 온도 여러 개를 리스트에 저장하기
3. 리스트 길이(len)와 특정 위치 값 꺼내기(인덱싱)
4. 설비 1대의 정보(설비번호·공정명·온도)를 딕셔너리로 표현하기
5. 여러 설비를 '딕셔너리의 리스트'로 표현해보기
