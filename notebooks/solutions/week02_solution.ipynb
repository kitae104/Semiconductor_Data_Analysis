# 2차시 정답 Notebook — 리스트와 딕셔너리

## 1단계. 리스트 만들기

```python
temperatures = [295.4, 298.1, 312.6, 301.0, 296.8]
print(temperatures)
```

## 2단계. 인덱싱으로 값 꺼내기

```python
print(temperatures[0])
print(temperatures[2])
print(temperatures[-1])
```

## 3단계. 슬라이싱

```python
print(temperatures[1:3])
```

**질문 답**: `temperatures[1:3]`은 인덱스 1, 2(298.1, 312.6)만 포함하고, `temperatures[1:4]`는 인덱스 1, 2, 3(298.1, 312.6, 301.0)까지 포함한다 — 끝 인덱스는 항상 결과에서 빠진다.

## 4단계. 길이와 추가

```python
print(len(temperatures))
temperatures.append(303.2)
print(temperatures)
```

## 5단계. 최댓값·최솟값·평균 계산하기

```python
highest = max(temperatures)
lowest = min(temperatures)
average = sum(temperatures) / len(temperatures)

print(f"최댓값: {highest}, 최솟값: {lowest}, 평균: {round(average, 1)}")
```

## 6단계. 딕셔너리 만들기

```python
equipment = {
    "설비번호": "EQ-01",
    "공정명": "식각",
    "온도_섭씨": 301.2,
}
print(equipment)
```

## 7단계. 키로 값 접근하기

```python
print(equipment["설비번호"])
print(equipment["온도_섭씨"])
```

## 8단계. 값 수정하기 / 새 키 추가하기

```python
equipment["온도_섭씨"] = 305.5
equipment["압력_Pa"] = 1012.3
print(equipment)
```

## 9단계. 리스트 안에 딕셔너리 담기

```python
equipment_list = [
    {"설비번호": "EQ-01", "공정명": "식각", "온도_섭씨": 301.2},
    {"설비번호": "EQ-02", "공정명": "증착", "온도_섭씨": 308.7},
    {"설비번호": "EQ-03", "공정명": "포토", "온도_섭씨": 297.5},
]

print(len(equipment_list))
print(equipment_list[1]["공정명"])
```

## 10단계(도전). 종합 실습 — 나만의 설비 목록 만들기

```python
my_equipment = [
    {"설비번호": "EQ-04", "공정명": "세정", "온도_섭씨": 299.0},
    {"설비번호": "EQ-05", "공정명": "산화", "온도_섭씨": 302.4},
]

print(f"{my_equipment[0]['설비번호']} 설비는 {my_equipment[0]['공정명']} 공정입니다.")
print(f"{my_equipment[1]['설비번호']} 설비는 {my_equipment[1]['공정명']} 공정입니다.")
```

## 11단계. 오늘의 학습을 한 문장으로 정리하기

> 리스트는 순서로 값을 구분하고, 딕셔너리는 이름표(키)로 값을 구분한다. 리스트 안에 딕셔너리를 담으면 여러 설비 정보를 한 번에 표현할 수 있다.

## 12단계. AI에게 질문하며 더 알아보기
오늘 배운 리스트·딕셔너리에 대해 AI 챗봇에게 질문해 이해를 넓혀봅니다.

질문 예시:
- "파이썬 리스트와 튜플은 어떻게 다른가요?"
- "딕셔너리에서 키(key)로 쓸 수 있는 값과 쓸 수 없는 값은 무엇인가요?"
- "딕셔너리 안에 리스트를 담을 수도 있나요?"
- "리스트의 append()와 insert()는 어떻게 다른가요?"

## 13단계. AI에게 코드 생성 요청하고 직접 실행해보기
프롬프트 예시: "설비 정보 딕셔너리가 여러 개 담긴 리스트에서, 온도가 가장 높은 설비 하나만
찾아주는 코드를 만들어줘."

아래는 AI가 생성해줄 수 있는 예시 코드와 실행 결과입니다.

```python
hottest = max(equipment_list, key=lambda eq: eq["온도_섭씨"])
print(f"가장 온도가 높은 설비는 {hottest['설비번호']}이고, 온도는 {hottest['온도_섭씨']}도입니다.")
```

## 14단계. 연습문제 — 압력 리스트와 새 설비 딕셔너리

```python
pressures = [1010.2, 1005.4, 1022.9, 998.7, 1011.0]

highest_p = max(pressures)
lowest_p = min(pressures)
average_p = sum(pressures) / len(pressures)

print(f"최댓값: {highest_p}, 최솟값: {lowest_p}, 평균: {round(average_p, 1)}")
```

### 도전. 새 설비 딕셔너리에 압력 추가하기

```python
eq04 = {
    "설비번호": "EQ-04",
    "공정명": "세정",
    "온도_섭씨": 299.9,
}

eq04["압력_Pa"] = average_p
print(eq04)
```

## 15단계. AI로 재미있는 미니 프로그램 만들기 🎉 — 오늘의 럭키 설비 뽑기
프롬프트 예시: "설비 정보 딕셔너리가 여러 개 담긴 리스트(equipment_list)에서, random.choice()로
하나를 무작위로 뽑아서 '오늘의 럭키 설비'라는 제목과 함께 설비번호·공정명·온도를 예쁘게 출력하는
파이썬 코드를 만들어줘."

아래는 AI가 생성해줄 수 있는 예시 코드와 실행 결과입니다(`eq04`를 `equipment_list`에 더해 네 개
설비 중에서 뽑히도록 구성했습니다).

```python
import random

random.seed(7)

equipment_list = [
    {"설비번호": "EQ-01", "공정명": "식각", "온도_섭씨": 301.2},
    {"설비번호": "EQ-02", "공정명": "증착", "온도_섭씨": 308.7},
    {"설비번호": "EQ-03", "공정명": "포토", "온도_섭씨": 297.5},
    eq04,
]

lucky = random.choice(equipment_list)
print("🎉 오늘의 럭키 설비 🎉")
print(f"설비번호: {lucky['설비번호']}")
print(f"공정명: {lucky['공정명']}")
print(f"온도: {lucky['온도_섭씨']}도")
```
