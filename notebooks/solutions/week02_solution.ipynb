# 2차시 실습 — 리스트와 딕셔너리 (정답)

쉬운 예제 값은 지어낸 값, 반도체 값은 교육용 가상 데이터입니다. 정상 온도 범위 295~305도는 학습용 기준입니다.

## 1단계. 리스트 만들기와 인덱스

```python
week_temps = [21.5, 23.0, 30.5, 24.0, 22.5, 25.0, 23.5]   # 월~일 낮 기온
print(week_temps)
```

```python
print(week_temps[0])    # 첫 번째 값(월요일)
print(week_temps[2])    # 세 번째 값(수요일)
print(week_temps[-1])   # 마지막 값(일요일)
```

```python
print(week_temps[1])    # 두 번째 값(첫 번째가 아니다!)
# print(week_temps[7])  # 칸은 0~6번뿐 → IndexError
```

## 2단계. 슬라이싱으로 여러 개 꺼내기

```python
print(week_temps[1:3])   # 1번 칸부터 3번 칸 앞까지(화·수)
print(week_temps[:5])    # 처음부터 5번 칸 앞까지(평일)
print(week_temps[5:])    # 5번 칸부터 끝까지(주말)
```

## 3단계. 개수 세기와 값 추가하기 — len(), append()

```python
cart = ["우유", "빵", "계란"]
print(len(cart))
```

```python
cart.append("사과")
print(cart)
print(len(cart))
```

```python
new_cart = []
new_cart.append("바나나")
new_cart.append("두부")
print(new_cart)
```

## 4단계. 최고·최저·평균 구하기 — max(), min(), sum()

```python
print(max(week_temps))
print(min(week_temps))
print(sum(week_temps))
```

```python
average = sum(week_temps) / len(week_temps)
print(f"이번 주 평균 기온은 {average:.1f}도입니다.")
```

## 5단계. 딕셔너리 만들고 키로 꺼내기

```python
student = {
    "이름": "지민",
    "나이": 15,
    "반": 3,
}
print(student)
```

```python
print(student["이름"])
print(student["반"])
# print(student["혈액형"])   # 없는 이름표 → KeyError
```

## 6단계. 값 바꾸기와 키 추가하기

```python
student["나이"] = 16           # 있던 키 → 값 바꾸기
student["동아리"] = "과학반"   # 없던 키 → 새로 추가
print(student)
```

```python
print(f"{student['이름']} 학생은 {student['나이']}살, 동아리는 {student['동아리']}입니다.")
```

## 7단계. 딕셔너리의 리스트 — 표의 한 행 = 딕셔너리 하나

```python
friends = [
    {"이름": "지민", "나이": 15, "동네": "서울"},
    {"이름": "서준", "나이": 16, "동네": "부산"},
    {"이름": "하은", "나이": 15, "동네": "대전"},
]
print(len(friends))
print(friends[0])
```

```python
print(friends[0]["나이"])   # 0번 행의 나이 칸
print(friends[1]["이름"])   # 1번 행의 이름 칸
```

```python
friend = friends[1]
print(f"{friend['이름']}: {friend['나이']}살, {friend['동네']}에 산다")
```

## 8단계. 반도체 데이터 실습 — 설비 온도 리스트

### 문제 1 — 인덱스로 꺼내기

```python
temperatures = [295.4, 298.1, 312.6, 301.0, 296.8]
print(temperatures)
```

```python
print(temperatures[0])    # 첫 번째 값
print(temperatures[2])    # 세 번째 값
print(temperatures[-1])   # 마지막 값
```

### 문제 2 — 슬라이싱

```python
print(temperatures[1:3])   # 1번 칸부터 3번 칸 앞까지
print(temperatures[:2])    # 처음부터 2번 칸 앞까지
print(temperatures[2:])    # 2번 칸부터 끝까지
```

### 문제 3 — 개수 세기와 추가

```python
print(len(temperatures))
```

```python
temperatures.append(303.2)
print(temperatures)
print(len(temperatures))
```

### 문제 4 — 최고·최저·평균

```python
temperatures = [295.4, 298.1, 312.6, 301.0, 296.8, 303.2]   # 하루 6번 잰 온도
print(len(temperatures))
```

```python
print(max(temperatures))
print(min(temperatures))
print(sum(temperatures))
```

```python
average = sum(temperatures) / len(temperatures)
print(f"평균 온도는 {average:.1f}도입니다.")
```

### 문제 5 — EQ-03 설비 온도 7건 요약

```python
eq03_temps = [303.8, 303.4, 292.2, 308.5, 307.5, 298.5, 304.0]
print(max(eq03_temps))
print(min(eq03_temps))
```

```python
average = sum(eq03_temps) / len(eq03_temps)
print(f"EQ-03 평균 온도는 {average:.1f}도입니다.")
```

## 9단계. 반도체 데이터 실습 — 설비 한 대의 기록(딕셔너리)

### 문제 6 — 딕셔너리 만들고 꺼내기

```python
equipment = {
    "설비번호": "EQ-01",
    "공정명": "식각",
    "온도_섭씨": 301.2,
}
print(equipment)
```

```python
print(equipment["설비번호"])
print(equipment["온도_섭씨"])
# print(equipment["압력_Pa"])   # 없는 이름표 → KeyError
```

### 문제 7 — 값 바꾸기와 키 추가

```python
equipment["온도_섭씨"] = 305.5   # 있던 키 → 값 바꾸기
equipment["합격여부"] = 1        # 없던 키 → 새로 추가(1 = 합격)
print(equipment)
```

```python
print(f"{equipment['설비번호']} 설비는 {equipment['공정명']} 공정, 온도는 {equipment['온도_섭씨']}도입니다.")
```

## 10단계. 반도체 데이터 실습 — 표 데이터(딕셔너리의 리스트)

### 문제 8 — 설비 3대의 표

### 문제 9 — 오늘 데이터 한 행 읽기

```python
records = [
    {"로트번호": "LOT-0122", "공정명": "식각", "설비번호": "EQ-02", "온도_섭씨": 302.5, "합격여부": 1},
    {"로트번호": "LOT-0097", "공정명": "포토", "설비번호": "EQ-04", "온도_섭씨": 308.7, "합격여부": 1},
    {"로트번호": "LOT-0026", "공정명": "산화", "설비번호": "EQ-03", "온도_섭씨": 303.8, "합격여부": 1},
]
print(len(records))
```
