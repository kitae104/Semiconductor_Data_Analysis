## 1단계. 비교 연산자와 if
완성된 코드입니다. 실행해서 결과를 확인하세요.

**주의**: `=`는 "저장하라", `==`는 "같은가?"입니다.

```python
score = 85
subject = "수학"

print(score > 80)
print(score <= 80)
print(subject == "수학")
print(subject != "수학")
```

```python
if score > 80:
    print("80점을 넘었습니다!")
print("채점 끝")
```

## 2단계. if/else와 if/elif/else로 나누기
위에서부터 묻다가 **처음 참인 곳에서 멈춥니다**.

```python
body_temp = 36.5

if 36.0 <= body_temp <= 37.5:
    print("정상 체온입니다.")
else:
    print("다시 재 보세요.")
```

```python
body_temp = 35.4

if body_temp > 37.5:
    print("높음")
elif body_temp >= 36.0:
    print("정상")
else:
    print("낮음")
```

## 3단계. for로 하나씩 꺼내 개수 세기
**주의**: 세는 변수(`= 0` 줄)는 반복문 **밖**에 둡니다.

```python
scores = [72, 85, 90, 64, 88]

for score in scores:
    print(score)
```

```python
pass_count = 0

for score in scores:
    if score >= 80:
        pass_count = pass_count + 1

print(f"80점 이상인 학생은 {pass_count}명입니다.")
```

## 4단계. 누적 합으로 합계와 평균 구하기
`total += money`는 `total = total + money`의 줄임말입니다.

```python
allowance = [5000, 3000, 7000, 2000]

total = 0

for money in allowance:
    total += money

average = total / len(allowance)
print(f"합계 {total}원, 평균 {average}원")
```

## 5단계. 딕셔너리의 리스트에서 골라내기
매번 딕셔너리 하나가 통째로 들어옵니다. 조건에 맞을 때만 `append()`로 담습니다.

```python
students = [
    {"이름": "민지", "반": "1반", "점수": 85},
    {"이름": "준호", "반": "2반", "점수": 72},
    {"이름": "서연", "반": "1반", "점수": 93},
]

for student in students:
    print(student["이름"], student["점수"])
```

```python
high_students = []

for student in students:
    if student["점수"] > 80:
        high_students.append(student["이름"])

print(f"80점을 넘은 학생: {high_students}")
```

## 6단계. 함수 만들기 — 평균 함수와 합격 판정 함수
3단계의 `scores`를 다시 씁니다(건너뛰었다면 `NameError: name 'scores' is not defined`). 완성된 코드입니다.

```python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)

print(calculate_average(scores))
```

```python
def judge_score(score):
    if score >= 80:
        return "합격"
    else:
        return "불합격"

print(judge_score(85))
print(judge_score(64))
```

```python
for score in scores:
    result = judge_score(score)
    print(f"{score}점 → {result}")
```

## 7단계. 함께 쓰기 — 학생 명단에서 불합격 판정 세기
5단계 `students`와 6단계 `judge_score`를 다시 씁니다. 완성된 코드입니다.

```python
fail_count = 0

for student in students:
    if judge_score(student["점수"]) == "불합격":
        fail_count += 1

print(f"불합격 판정을 받은 학생: {fail_count}명")
```
