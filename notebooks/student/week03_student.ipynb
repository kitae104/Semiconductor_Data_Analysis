# 3차시 학생용 실습 — 리스트·딕셔너리로 배우는 조건문, 반복문, 함수

오늘의 질문: **리스트에 담긴 여러 설비 정보를 하나씩 살펴보며, 조건에 맞는 것만 컴퓨터가 골라내게 할 수 있을까?**

> 🚧 이 노트북은 v2 커리큘럼 구성이 확정된 직후 작성된 **개요(skeleton)**입니다. 다룰 단계와 순서는 정해졌지만, 구체적인 코드·설명·빈칸 문항은 아직 채워지지 않았습니다. 구성이 최종 확정되면 이어서 상세 내용을 작성합니다.

## 다룰 실습 단계(초안)

1. 온도 하나를 조건문으로 정상/비정상 판정하기
2. for로 리스트 전체를 순회하며 정상 개수 세기
3. '딕셔너리의 리스트'를 순회하며 조건에 맞는 설비만 골라내기
4. 평균 계산을 함수로 만들기
5. 합격/불합격 판정을 함수로 만들어 여러 로트에 반복 적용하기
