## 1단계. 데이터 준비하기
도구·한글 글꼴 설정 후 아이스크림 가게 기록을 읽고 날짜를 바꿉니다(실행만).

**주의**: 글꼴 두 줄을 빼면 한글이 □로 나옵니다(Mac은 `"AppleGothic"`).

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 글꼴(Windows)
plt.rcParams["axes.unicode_minus"] = False      # 음수 부호 깨짐 방지
```

```python
ice = pd.read_csv("../../data/weekly/week06/week06_easy_icecream_daily.csv", encoding="utf-8-sig")
ice.head()
```

```python
ice["날짜"] = pd.to_datetime(ice["날짜"])
ice["날짜"].head(3)
```

## 2단계. 선 그래프 — 날짜별 판매량
**주의**: 흔들림 전체가 조금씩 위로 가는지를 보세요.

```python
plt.plot(ice["날짜"], ice["판매량_개"])
plt.xticks(rotation=45)   # 날짜 글자를 45도 기울여 겹치지 않게
plt.xlabel("날짜")
plt.ylabel("판매량_개")
plt.show()
```

## 3단계. 막대그래프 — 날씨별 날 수
`value_counts()`로 세고, 그 숫자를 막대로 그립니다.

```python
counts = ice["날씨"].value_counts()
print(counts)
```

```python
plt.bar(counts.index, counts)
plt.xlabel("날씨")
plt.ylabel("날 수")
plt.show()
```

## 4단계. 히스토그램 — 판매량이 몰린 구간
`bins=`는 칸의 개수. **주의**: 가로축은 이름이 아니라 숫자 구간입니다.

```python
plt.hist(ice["판매량_개"], bins=5)
plt.xlabel("판매량_개")
plt.ylabel("날 수")
plt.show()
```

## 5단계. 산점도 — 기온과 판매량
점 하나 = 하루. **주의**: 함께 움직여도 "원인"이라는 뜻은 아닙니다.

```python
plt.scatter(ice["최고기온_섭씨"], ice["판매량_개"])
plt.xlabel("최고기온_섭씨")
plt.ylabel("판매량_개")
plt.show()
```

## 6단계. 박스플롯 — 날씨별 판매량
**주의**: 영어 제목 "Boxplot grouped by …"는 자동으로 붙습니다(오류 아님).

```python
ice.boxplot(column="판매량_개", by="날씨")
plt.show()
```

## 7단계. 숫자로 다시 확인하기 — 날씨별 숫자 요약
`groupby`는 7차시에 배웁니다(실행해서 결과만 읽기). `round(1)` = 소수 첫째 자리까지. `std` = 퍼짐, `25%`·`50%`·`75%` = 상자 아래 끝·가운데 선·위 끝.

```python
print(ice.groupby("날씨")["판매량_개"].describe().round(1))
```
