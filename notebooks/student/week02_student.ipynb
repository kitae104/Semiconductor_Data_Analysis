## 1단계. 리스트 만들기와 인덱스
**주의**: 번호는 **0부터** 셉니다. 첫 번째 `[0]`, 세 번째 `[2]`, 마지막 `[-1]`.

```python
week_temps = [21.5, 23.0, 30.5, 24.0, 22.5, 25.0, 23.5]   # 월~일 낮 기온
print(week_temps)
```

```python
print(week_temps[0])    # 첫 번째 값(월요일)
print(week_temps[2])    # 세 번째 값(수요일)
print(week_temps[-1])   # 마지막 값(일요일)
```

둘째 줄 `#`을 지우면 `IndexError`가 납니다(확인 후 다시 붙이기).

```python
print(week_temps[1])    # 두 번째 값(첫 번째가 아니다!)
# print(week_temps[7])  # 칸은 0~6번뿐 → IndexError
```

## 2단계. 슬라이싱으로 여러 개 꺼내기
**주의**: 끝 번호 칸은 들어가지 않습니다(개수 = 끝 - 시작).

```python
print(week_temps[1:3])   # 1번 칸부터 3번 칸 앞까지(화·수)
print(week_temps[:5])    # 처음부터 5번 칸 앞까지(평일)
print(week_temps[5:])    # 5번 칸부터 끝까지(주말)
```

## 3단계. 개수 세기와 값 추가하기 — len(), append()
**주의**: `append` 셀은 **한 번만** 실행하세요(다시 실행하면 값이 또 붙습니다).

```python
cart = ["우유", "빵", "계란"]
print(len(cart))
```

```python
cart.append("사과")
print(cart)
print(len(cart))
```

```python
new_cart = []
new_cart.append("바나나")
new_cart.append("두부")
print(new_cart)
```

## 4단계. 최고·최저·평균 구하기 — max(), min(), sum()
평균 = `sum()` ÷ `len()`. 평균과 최고 기온을 비교해 보세요.

```python
print(max(week_temps))
print(min(week_temps))
print(sum(week_temps))
```

```python
average = sum(week_temps) / len(week_temps)
print(f"이번 주 평균 기온은 {average:.1f}도입니다.")
```

## 5단계. 딕셔너리 만들고 키로 꺼내기
**주의**: 글자 값과 키는 따옴표로 감싸고, 숫자는 그대로 씁니다. 마지막 줄 `#`을 지우면 `KeyError`가 납니다.

```python
student = {
    "이름": "지민",
    "나이": 15,
    "반": 3,
}
print(student)
```

```python
print(student["이름"])
print(student["반"])
# print(student["혈액형"])   # 없는 이름표 → KeyError
```

## 6단계. 값 바꾸기와 키 추가하기
**주의**: f-string 안쪽 키는 작은따옴표(`'이름'`)로 씁니다. 겹치면 `SyntaxError`가 납니다.

```python
student["나이"] = 16           # 있던 키 → 값 바꾸기
student["동아리"] = "과학반"   # 없던 키 → 새로 추가
print(student)
```

```python
print(f"{student['이름']} 학생은 {student['나이']}살, 동아리는 {student['동아리']}입니다.")
```

## 7단계. 딕셔너리의 리스트 — 표의 한 행 = 딕셔너리 하나
**주의**: `[행 번호]` 먼저, `[키]` 다음. 순서를 바꾸면 `TypeError`가 납니다.

```python
friends = [
    {"이름": "지민", "나이": 15, "동네": "서울"},
    {"이름": "서준", "나이": 16, "동네": "부산"},
    {"이름": "하은", "나이": 15, "동네": "대전"},
]
print(len(friends))
print(friends[0])
```

```python
print(friends[0]["나이"])   # 0번 행의 나이 칸
print(friends[1]["이름"])   # 1번 행의 이름 칸
```

```python
friend = friends[1]
print(f"{friend['이름']}: {friend['나이']}살, {friend['동네']}에 산다")
```
