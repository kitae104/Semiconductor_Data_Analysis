# 2차시 학생용 실습 — 리스트와 딕셔너리

오늘의 질문: **여러 개의 측정값을 한 번에 담고, 각 값에 이름표까지 붙이려면 어떻게 해야 할까?**

> 🚧 이 노트북은 v2 커리큘럼 구성이 확정된 직후 작성된 **개요(skeleton)**입니다. 다룰 단계와 순서는 정해졌지만, 구체적인 코드·설명·빈칸 문항은 아직 채워지지 않았습니다. 구성이 최종 확정되면 이어서 상세 내용을 작성합니다.

## 다룰 실습 단계(초안)

1. 온도 여러 개를 리스트에 담기
2. 리스트에서 최댓값·최솟값·평균 직접 계산하기
3. 설비 1대의 정보(설비번호·공정명·온도)를 딕셔너리로 표현하기
4. 여러 설비를 '딕셔너리의 리스트'로 표현하기
5. 리스트 안 딕셔너리에서 특정 설비의 값 꺼내기
