강의의 AI 예시 코드는 로트 이름(LOT-A~D)을 지어내고 불합격 확률 4개를 직접 적어 넣었습니다. 같은 반복문에 오늘 만든 `new_lots`의 이름표와 `불합격_확률` 열을 넣으면 새 로트 9건 전체의 위험도를 매길 수 있습니다. 위험도 기준(0.3, 0.15)은 여전히 AI가 임의로 정한 값입니다.

```
# AI가 만들어 준 위험도 등급 코드를 이 셀에 붙여넣고 실행합니다.
# 실행하면 로트별 불합격 확률과 위험도(높음/보통/낮음)가 출력되는지 확인해 보세요.
# 여기에 AI가 생성한 코드를 붙여넣고 실행해보세요.
```

```
# AI 예시의 지어낸 로트 이름 대신, 새 로트 9건에 "공정명-설비번호" 형태의 이름표를 붙입니다(글자 열끼리 +로 이어 붙임).
# 실행하면 이름표 9개가 보입니다 — 로트 번호가 따로 없어서 같은 이름표가 두 번 나오는 곳이 있는지 확인해 보세요.
labels = new_lots["공정명"] + "-" + new_lots["설비번호"]
labels  # (중간 확인)
```

```
# 위 AI 예시 코드의 반복문을 그대로 쓰되, lots·fail_proba 대신 labels와 4단계의 불합격_확률 열을 넣습니다.
# 실행하면 새 로트 9건이 확률이 높은 순서로 위험도와 함께 출력됩니다 — "높음"인 로트가 5단계의 두 로트와 같은지 확인해 보세요.
for lot, p in sorted(zip(labels, new_lots["불합격_확률"]), key=lambda x: x[1], reverse=True):
    if p >= 0.3:
        risk = "높음"
    elif p >= 0.15:
        risk = "보통"
    else:
        risk = "낮음"
    print(f"{lot}: 불합격 확률 {p:.1%} → 위험도 {risk}")
```
