## 1단계. 9차시 모델 다시 만들기
차례로 실행하며 `(116, 6)` → `(116, 4)` → `92 24`를 확인합니다.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

easy_df = pd.read_csv("../../data/weekly/week10/week10_easy_outdoor_class_train.csv", encoding="utf-8-sig").dropna()
print(easy_df.shape)
```

```python
easy_feature_cols = ["최고기온_섭씨", "습도_pct", "풍속_m_s", "강수량_mm"]
easy_X = easy_df[easy_feature_cols]   # 단서(Feature) 4개
easy_y = easy_df["진행여부"]            # 정답(Label): 1 = 진행, -1 = 취소
print(easy_X.shape)
```

```python
easy_X_train, easy_X_test, easy_y_train, easy_y_test = train_test_split(
    easy_X, easy_y, test_size=0.2, random_state=42, stratify=easy_y
)
print(len(easy_X_train), len(easy_X_test))
```

```python
easy_model = DecisionTreeClassifier(max_depth=4, random_state=42)
easy_model.fit(easy_X_train, easy_y_train)
```

## 2단계. 모델 평가하기 — 정확도·기준선·혼동행렬
정확도 → 진행·취소 개수 → 기준선 → 혼동행렬 순서로 실행합니다.

- 빈칸: "진행"을 뜻하는 숫자(**1 = 진행**)

```python
from sklearn.metrics import accuracy_score

easy_pred = easy_model.predict(easy_X_test)
easy_accuracy = accuracy_score(easy_y_test, easy_pred)
print(f"모델 정확도: {easy_accuracy:.1%}")
```

```python
print(easy_y_test.value_counts())
```

```python
from sklearn.metrics import confusion_matrix

easy_cm = confusion_matrix(easy_y_test, easy_pred, labels=[1, -1])
print(easy_cm)
```

## 4단계. 확률까지 확인하기 — predict_proba()
`classes_`가 `[-1  1]`이면 **0번 열 = 취소(-1)일 확률**입니다. `easy_proba[:, 0]`은 "모든 줄(`:`)의 0번 칸"입니다.

```python
easy_proba = easy_model.predict_proba(easy_X_new)
print(easy_model.classes_)
print(easy_proba.round(3))
```
