## 2단계. 빈칸 세고 지우기 — isna(), dropna()

`isna().sum()`은 열마다 빈칸(NaN) 개수를 셉니다. `dropna()`는 빈칸이 하나라도 있는 행을 지웁니다. **주의**: 지운 결과를 `students =`로 다시 저장해야 남습니다.

```python
print(students.isna().sum())
```

```python
students = students.dropna()
print(students.shape)
```

## 3단계. 중복 행 찾고 지우기 — duplicated(), drop_duplicates()

```python
print(students.duplicated().sum())
```

```python
students = students.drop_duplicates()
print(students.shape)
```

## 4단계. 튀는 값 찾기 — 범위 조건

사람 키가 250cm를 넘는 일은 거의 없습니다. 오늘은 250cm를 넘으면 잘못 적은 값으로 보고, 4차시에 배운 조건 검색으로 찾아봅니다.

```python
tall = students[students["키_cm"] > 250]
print(tall)
```

## 5단계. 튀는 값 빼고 전후 비교하기

```python
students = students[students["키_cm"] <= 250]
print(students.shape)
```

```python
after_height = round(students["키_cm"].mean(), 1)
print("정리 전 평균 키:", before_height)
print("정리 후 평균 키:", after_height)
```

> week05_dirty_process_data.csv를 pandas로 읽어서(encoding="utf-8-sig"), 열마다 빈칸 개수와 빈칸 비율(%)을 한 줄씩 출력하는 짧은 코드를 만들어 줘.

아래는 받은 코드의 예시입니다. `for`는 3차시에 배운 반복문입니다. 실행해서 6단계 `isna().sum()` 결과와 같은지 확인하세요.

```python
import pandas as pd

data = pd.read_csv("week05_dirty_process_data.csv", encoding="utf-8-sig")

for col in data.columns:
    count = data[col].isna().sum()
    rate = count / len(data) * 100
    print(f"{col}: 빈칸 {count}개 ({rate:.1f}%)")
```
