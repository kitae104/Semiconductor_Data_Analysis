## 1단계. 데이터 읽고 숫자 요약 보기
반 학생 기록을 읽고 요약합니다(완성된 코드).

```python
import pandas as pd

students = pd.read_csv("../../data/weekly/week05/week05_easy_class_records.csv", encoding="utf-8-sig")
print(students.shape)
```

```python
print(students.describe())
```

**주의**: 아래 "정리 전 값" 셀은 정리 전에 **한 번만** 실행합니다.

```python
before_students = len(students)
before_height = students["키_cm"].mean()
before_weight = students["몸무게_kg"].mean()

print(f"정리 전: {before_students}명, 평균 키 {before_height:.1f}, 평균 몸무게 {before_weight:.1f}")
```

## 2단계. 결측값 세고 처리하기
빈칸 세기 → 반이 빈 행 지우기 → 키·몸무게 빈칸을 중앙값으로 채우기(완성된 코드).

```python
print(students.isna().sum())
```

```python
students = students.dropna(subset=["반"])
print(students.shape)
```

```python
for col in ["키_cm", "몸무게_kg"]:
    median_value = students[col].median()
    students[col] = students[col].fillna(median_value)

print(students.isna().sum())
```

## 3단계. 중복 행 찾고 지우기
똑같은 행 세기 → 들여다보기 → 지우기(완성된 코드).

```python
print(students.duplicated().sum())
```

```python
dup = students[students.duplicated()]
print(dup[["이름", "반", "키_cm", "몸무게_kg"]])
```

```python
students = students.drop_duplicates()
print(students.shape)
```

## 4단계. 이상값 찾고 들여다보기
키 250cm 초과 기록을 찾고 다른 열과 함께 봅니다(완성된 코드).

```python
tall = students[students["키_cm"] > 250]
print("키 250 초과:", len(tall), "명")
```

```python
print(tall[["이름", "반", "키_cm", "몸무게_kg"]])
```

## 5단계. 이상값 지우기
조건의 **반대**(키 250 이하)만 남깁니다(완성된 코드).

```python
students = students[students["키_cm"] <= 250]
print(students.shape)
```

## 6단계. 정리 전후 비교하기
정리 전 값과 지금 값을 나란히 출력합니다(완성된 코드).

```python
after_height = students["키_cm"].mean()
after_weight = students["몸무게_kg"].mean()

print(f"학생 수: {before_students} → {len(students)}")
print(f"평균 키: {before_height:.1f} → {after_height:.1f}")
print(f"평균 몸무게: {before_weight:.1f} → {after_weight:.1f}")
```
